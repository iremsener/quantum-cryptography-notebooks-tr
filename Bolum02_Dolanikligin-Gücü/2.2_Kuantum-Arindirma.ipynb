## 2.2 Kuantum Arındırma (Purifications)

Daha büyük bir birleşik sistemin tanımı verildiğinde, bir alt sistemin (parçanın) kuantum durumunu yalnız bırakıp açıklamamızı sağlayan **kısmi iz (partial trace)** işlemini öğrenmiştik. Orada çok şaşırtıcı bir şey fark etmiştik: Bileşke sistemin kendisi tamamen saf (pure) bir kuantum durumunda olsa bile, onun içinden bir alt sistemi çektiğimizde elde ettiğimiz indirgenmiş durum bir karma durum (mixed state) çıkabiliyordu. Bu durum, büyük sistemin içinde kuantum dolanıklığının var olduğunun en kesin imzasıydı.

Peki, bu süreci tamamen **tersine çevirmek** mümkün müdür? 

Yani, $A$ sistemi üzerinde tanımlı klasik belirsizlikler barındıran karma bir $\rho_A$ yoğunluk matrisi verildiğinde; öyle saf bir $|\Psi\rangle_{AB}$ birleşik durumu bulabilir miyiz ki, Bob'un uzayını yok ettiğimizde ($\text{tr}_B$) geriye tam olarak bizim başlangıçtaki $\rho_A$ matrisimiz kalsın? Yanıtın evet olduğu ortaya çıkıyor! Kuantum literatüründe bu matematiksel genişletme işlemine **arındırma (purification)** adı verilir.

> 🟦 **Tanım 2.2.1 — Kuantum Arındırma.** Rastgele bir $\rho_A$ yoğunluk matrisini ele alalım. Eğer saf bir $|\Psi\rangle_{AB}$ birleşik durumu, Bob üzerinden kısmi iz alındığında Alice'in orijinal matrisini geri veriyorsa; yani **$\text{tr}_B(|\Psi\rangle\langle\Psi|_{AB}) = \rho_A$** şartı sağlanıyorsa, bu $|\Psi\rangle_{AB}$ durumuna $\rho_A$ matrisinin bir **arındırılmış hali (purification)** denir.


Herhangi bir rastgele $\rho_A$ yoğunluk matrisinin kağıt üzerinde nasıl adım adım arındırılabileceğini inceleyelim. 

**Adım 1:** İlk olarak, Hermityen olan $\rho_A$ matrisini spektral ayrışım yöntemiyle kendi özdurumları cinsinden köşegenleştirelim (bir karışım olarak yazalım):

$$\rho_A = \sum_{j=1}^{d_A} \lambda_j | \phi_j \rangle\langle \phi_j | \quad \text{--- (2.14)}$$

Burada $\lambda_j$ değerleri matrisin negatif olmayan özdeğerleridir ve toplamları 1'e eşittir ($\sum \lambda_j = 1$). Bu ifade bize sistemin $\lambda_j$ olasılığıyla $|\phi_j\rangle$ durumunda hazırlandığını söyler.

**Adım 2:** Şimdi, Alice'in uzayının boyutuna eşit boyutta ($d_B = d_A$) olan ve standart taban elemanlarını $\{|j\rangle_B\}$ şeklinde göstereceğimiz **hayali (imaginary) bir B kuantum sistemi** tasarlayalım. Bu hayali sistemi denkleme ortak ederek, birleşik uzayda şu saf kuantum durumunu inşa edelim:

$$|\Psi\rangle_{AB} = \sum_{j=1}^{d_A} \sqrt{\lambda_j} | \phi_j \rangle_A \otimes | j \rangle_B \quad \text{--- (2.15)}$$

Bu yeni durumun en harika özelliği şudur: Eğer Bob kendi hayali sistemi ($B$) üzerinde standart tabanda bir ölçüm gerçekleştirirse, $j$ sonucunu elde etme olasılığı tam olarak bizim orijinal özdeğerimiz olan $\lambda_j$ değerine eşit çıkar. Dahası, ölçüm sonrasında Alice'in elinde kalan durum tam olarak $|\phi_j\rangle\langle\phi_j|_A$ olur. 

Toplam olasılık teorisi altında Bob'un ölçüm sonuçlarını birleştirdiğimizde ($\text{tr}_B$ aldığımızda), karşımıza kuruşu kuruşuna bizim başlangıçtaki $\rho_A = \sum \lambda_j |\phi_j\rangle\langle\phi_j|$ matrisimiz çıkar. Böylece arındırma işleminin her zaman doğrulanabilir bir şekilde mümkün olduğu ispatlanmış olur.


Peki, bir karma durum için tanımlanabilecek arındırma durumu **benzersiz ve tek midir?** 

Yukarıdaki matematiksel inşayı yaparken, Bob'un hayali sistemi için tamamen rastgele bir standart taban $(\{|j\rangle_B\})$ seçimi yapmıştık. Oysa Bob'un uzayında seçilebilecek sonsuz farklı ortonormal taban kombinasyonu mevcuttur ve onların her biriyle tamamen geçerli, farklı bir arındırma durumu yazılabilir. 

Bu durum, kuantum mekaniğinde arındırma işlemlerinin **uniter bir serbestlik derecesine (unitary degree of freedom)** sahip olduğu anlamına gelir. Bu serbestliğin sınırlarını ve tüm geçerli arındırma durumlarının birbiriyle olan derin matematiksel ilişkisini tam olarak kavrayabilmek için, bir sonraki bölümde kuantum bilgi teorisinin en kullanışlı araçlarından biri olan **Schmidt Ayrışımı (Schmidt Decomposition)** konusunu öğreneceğiz.


### 2.2.1 Schmidt Ayrışımı (The Schmidt decomposition)

Bir önceki bölümde (2.15) numaralı denklemle kurduğumuz arındırma durumunun son derece özel ve simetrik bir formu vardı: Katsayılarının kareleri toplamı 1 olan, $A$ ve $B$ sistemlerine ait taban vektörlerinin doğrudan tensör çarpımlarının bir toplamı olarak ifade edilmişti. Gördüğümüz üzere bu biçim, alt sistemlerin indirgenmiş durumlarını kuruşu kuruşuna hesaplamayı aşırı derecede kolaylaştırıyordu.

Ancak ne yazık ki evrendeki her iki-parçalı saf kuantum durumu bize başlangıçta bu kadar temiz ve hazır bir biçimde verilmez! Örneğin, şu iki kübitlik $|\Psi\rangle_{AB}$ durumunu ele alalım:

$$|\Psi\rangle_{AB} = \frac{1}{\sqrt{2}}\Big( |0\rangle_A|0\rangle_B + |+\rangle_A|1\rangle_B \Big)$$

Buradaki $|0\rangle_A$ ve $|+\rangle_A$ durumları Alice'in uzayında tanımlıdır fakat **birbirine dik (orthogonal) değillerdir** ($\langle0|+\rangle = 1/\sqrt{2} \neq 0$). Durum böyle olunca, sistemin alt parçalarını yalnız bırakmak veya dolanıklık analizini yapmak matematiksel olarak oldukça sancılı bir hal alır. Peki, başlangıçta karmaşık görünen böyle bir durumu, iki taraf için de ortonormal tabanlar barındıran daha kullanışlı ve sade bir formda yeniden yazabilir miyiz? 

Yanıtın evet olduğu ortaya çıkıyor! Bizi bu pratik forma ulaştıran yapı kuantum mekaniğindeki meşhur **Schmidt Ayrışımı** teoremidir.


> 🟦 **Teorem 2.2.1 — Schmidt Ayrışımı (Schmidt Decomposition).** Sırasıyla $d_A$ ve $d_B$ boyutlarına sahip $A$ ve $B$ kuantum sistemlerini ele alalım ve sistemlerin minimum boyut sınırını $d = \min(d_A, d_B)$ olarak belirleyelim. Herhangi bir iki-parçalı saf $|\Psi\rangle_{AB}$ kuantum durumu, her iki taraf için de özel seçilmiş yeni tabanlar cinsinden her zaman şu formda ayrıştırılabilir:
> 
> $$|\Psi\rangle_{AB} = \sum_{i=1}^{d} \sqrt{\lambda_i} |u_i\rangle_A \otimes |v_i\rangle_B \quad \text{--- (2.16)}$$
> 
> Burada her bir $i$ için $\lambda_i \geq 0$ olup, $\{|u_i\rangle_A\}$ ve $\{|v_i\rangle_B\}$ kümeleri sırasıyla $A$ ve $B$ uzayları için **birebir ortonormal vektör kümeleridir**. 
> 
> Kuantum literatüründe buradaki $\sqrt{\lambda_i}$ katsayılarına **Schmidt katsayıları (Schmidt coefficients)**, elde edilen yeni $\{|u_i\rangle_A\}$ ve $\{|v_i\rangle_B\}$ taban vektörlerine ise **Schmidt vektörleri (Schmidt vectors)** adı verilir.


Schmidt Ayrışımı'nın kuantum bilgi teorisine getirdiği asıl devrimsel kolaylık şudur: Başlangıçta Alice ve Bob'un uzayları birbiriyle ne kadar karmaşık bir şekilde iç içe geçmiş veya dolanmış olursa olsun, bu teorem bize iki tarafın da kendi lokal laboratuvarlarında öyle akıllıca iki yeni taban ($\{|u_i\rangle_A\}$ ve $\{|v_i\rangle_B\}$) seçebileceğini söyler ki, birleşik durum tıpkı (2.16)'daki gibi tekil indisler üzerinden ($i=1$ için $11$, $i=2$ için $22$ gibi) kusursuz bir eşleşmeyle paralel bir forma kavuşur. 

Bu sayede, karmaşık çapraz terimlerin ($|01\rangle$ veya $|10\rangle$ gibi) tamamı matematiksel olarak elenir. Bir durumun Schmidt katsayılarının sayısı (sıfırdan farklı olanlar), o sistemin içinde ne kadar kuantum dolanıklığı barındırdığının en temel ve en saf ölçüsüdür!


```python
import numpy as np

# Temel kübit durumları
q0 = np.array([1, 0])
q1 = np.array([0, 1])
plus = np.array([1, 1]) / np.sqrt(2)

# Nottaki test durumunu oluşturalım: |Psi_AB> = 1/sqrt(2) * (|00> + |+1>)
# np.kron(,) -> |00>
# np.kron(,) -> |+1>
psi_ket = (np.kron(q0, q0) + np.kron(plus, q1)) / np.sqrt(2)

# 4x1 boyutundaki bu birleşik vektörü Alice ve Bob uzaylarına göre 2x2 boyutunda bir matrise dönüştürüyoruz
psi_matrix = psi_ket.reshape(2, 2)

# Lineer cebirdeki SVD (Singular Value Decomposition) ile Schmidt katsayılarını yalnız bırakıyoruz
# U -> Alice Schmidt vektörleri, V_dagger -> Bob Schmidt vektörleri, S -> Schmidt katsayıları
U, S, V_dagger = np.linalg.svd(psi_matrix)

print("Matematiksel Durum Vektörü Çıktısı:", psi_ket)
print("\nHesaplanan Schmidt Katsayıları (sqrt(lambda_i)):", S)
print(f"Katsayıların karelerinin toplamı: {np.sum(S**2):.1f} (1.0 olmalı)")
```

Schmidt Ayrışımı teoremine ait detaylı ispat adımlarına ders videolarından veya [NC01] (Nielsen & Chuang) kaynağının 2.5 nolu bölümünden ulaşabilirsiniz. İspatın temel fikri; birleşik durumu öncelikle standart tabanda $\sum \alpha_{jk} |j\rangle|k\rangle$ şeklinde yazmaya, ardından katsayılardan oluşan $\alpha_{jk}$ matrisine **Tekil Değer Ayrışımı (SVD)** uygulayarak Schmidt katsayılarını ($\sqrt{\lambda_i}$) ve iki tarafa ait Schmidt vektörlerini ($|u_i\rangle_A, |v_i\rangle_B$) yalnız bırakmaya dayanır.

Bu ayrışımın kuantum bilgi teorisinde çok çarpıcı sonuçları vardır:

1. **İndirgenmiş Durumların Kolayca Bulunması:** Schmidt formunda (2.16) verilen bir saf durumun kısmi izini almak aşırı derecede kolaydır. Doğrudan $\rho_A = \sum \lambda_i |u_i\rangle\langle u_i|_A$ ve $\rho_B = \sum \lambda_i |v_i\rangle\langle v_i|_B$ marjinal matrislerini elde ederiz.
2. **Spektrum Eşitliği:** En can alıcı gözlem, Alice ve Bob'un indirgenmiş durumlarının ($\rho_A$ ve $\rho_B$) **özdeğerlerinin (spektrumlarının) kuruşu kuruşuna tamamen aynı olmasıdır!** Bu özdeğerler, Schmidt katsayılarının kareleri olan $\lambda_i$ değerleridir. 
3. **Arındırma Şartı:** Bu durum bize şunu söyler: İki farklı karma durum $\rho_A$ ve $\rho_B$ verildiğinde, bunları ortak bir saf $|\Psi\rangle_{AB}$ durumunda birleştirebilmemiz (arındırabilmemiz) için gerek ve yeter şart, bu iki matrisin **aynı özdeğer kümesine (aynı spektruma)** sahip olmasıdır! Schmidt ayrışımı olmadan bu gerçeği kağıt üzerinde ispatlamak neredeyse imkansızdır.


Schmidt katsayıları matematiksel olarak tamamen **benzersiz (unique)** ve tek bir şekilde tanımlanmıştır; çünkü doğrudan indirgenmiş yoğunluk matrisinin özdeğerlerinin karekökleridir. Schmidt vektörleri ise (dejenerelik ve evre seçimi hariç) yine benzersizdir. 

Eğer bir özdeğere karşılık gelen özuzay 1 boyutlu ise, o Schmidt vektörü kesinlikle o özvektör olmak zorundadır. Ancak özuzay birden fazla boyuta sahipse, o alt uzay için herhangi bir ortonormal taban kümesini Schmidt vektörleri olarak seçebiliriz. Ayrıca uniter bir serbestlik derecesi olarak; (2.16) denklemindeki $|u_i\rangle$ vektörünü $e^{i\theta_i}$ ile çarparken, $|v_i\rangle$ vektörünü $e^{-i\theta_i}$ ile çarparak **evre serbestliğini (phase degree of freedom)** koruyabiliriz; bu durum bileşke tensör çarpım sonucunu zerre kadar değiştirmez.


Schmidt Ayrışımı'nın bize sunduğu en pratik araçlardan biri de, iki sistem arasındaki kuantum dolanıklığının miktarını doğrudan ölçebilmemizi sağlamasıdır. Bu doğrultuda tanımlanan ilk kaba ama oldukça kullanışlı ölçü, sıfırdan farklı Schmidt katsayılarının ($\sqrt{\lambda_j}$) toplam sayısıdır. Bu sayıya **Schmidt derecesi (Schmidt rank)** adı verilir.

> 🟦 **Tanım 2.2.2 — Schmidt Derecesi (Schmidt Rank).** Schmidt ayrışımı $|\Psi\rangle_{AB} = \sum_{i=1}^d \sqrt{\lambda_i} |a_i\rangle|b_i\rangle_B$ şeklinde verilen herhangi bir iki-parçalı saf kuantum durumu için Schmidt derecesi, **sıfırdan büyük ($\sqrt{\lambda_i} > 0$) olan Schmidt katsayılarının toplam sayısıdır**. Bu değer aynı zamanda matematiksel olarak $\text{rank}(\rho_A)$ ve $\text{rank}(\rho_B)$ değerlerine birebir eşittir.
> 
> *   **Schmidt Derecesi = 1 ise:** Durum basit bir çarpım durumudur (product state); yani içinde **hiç dolanıklık yoktur**.
> *   **Schmidt Derecesi > 1 ise:** Durum kesinlikle **dolanıktır (entangled)**.


```python
import numpy as np

# Temel kübit durumları
q0 = np.array([1, 0])
q1 = np.array([0, 1])

# Test için meşhur dolanık EPR durumunu kuralım: 1/sqrt(2) * (|00> + |11>)
epr_ket = (np.kron(q0, q0) + np.kron(q1, q1)) / np.sqrt(2)

# Vektörü 2x2 boyutunda bir matrise dönüştürüp Tekil Değer Ayrışımı (SVD) yapalım
psi_matrix = epr_ket.reshape(2, 2)
U, schmidt_coefficients, V_dagger = np.linalg.svd(psi_matrix)

# Sıfırdan farklı katsayıları sayarak Schmidt Derecesini (Schmidt Rank) bulalım
# Sayısal hassasiyet payı (1e-9) eklenmiştir
schmidt_rank = np.sum(schmidt_coefficients > 1e-9)

print("Hesaplanan Schmidt Katsayıları:", np.round(schmidt_coefficients, 4))
print("Sistemin Schmidt Derecesi (Schmidt Rank):", schmidt_rank)

# Tanım 2.2.2'ye göre dolanıklık kontrolü
if schmidt_rank > 1:
    print("ANALİZ SONUCU: Sistem DOLANIKTIR (Entangled).")
else:
    print("ANALİZ SONUCU: Sistem BAĞIMSIZDIR (Product State / Ayrılabilir).")
```

Schmidt katsayıları, Schmidt derecesine (rank) kıyasla iki sistem arasındaki dolanıklığı ölçmek için çok daha hassas ve ince bir yol sunar. **"Dolanıklık entropisi" (entropy of entanglement)** olarak adlandırılan bu doğal yöntem, Schmidt katsayılarının kareleri ($\lambda_i$) tarafından belirlenen olasılık dağılımının **Von Neumann entropisini** (veya klasik anlamda Shannon entropisini) hesaplamaya dayanır.

Eğer bu dağılımın entropisi tam olarak 0 ise, bu durum katsayılardan yalnızca bir tanesinin 1'e eşit olduğu ve sistemin dolanık olmadığı (çarpım durumu olduğu) anlamına gelir. Ancak entropi pozitif bir değer aldığı andan itibaren durumun dolanık olduğu kesinleşir. Bu ölçüm yöntemi, Schmidt derecesine kıyasla çok daha ayırt edicidir. Örneğin, aşağıdaki iki farklı kuantum durumunun içerdiği dolanıklık kalitesini birbirinden kolayca ayırabilir:

$$|\Psi\rangle = \frac{1}{\sqrt{2}}|00\rangle + \frac{1}{\sqrt{2}}|11\rangle \quad \text{ve} \quad |\phi\rangle = \sqrt{1 - \epsilon}|00\rangle + \sqrt{\epsilon}|11\rangle \quad \text{--- (2.17)}$$


Küçük bir $0 < \epsilon < 1/2$ parametresi için bu iki durumu incelediğimizde; her iki durumun da sıfırdan farklı ikişer katsayısı olduğu için **Schmidt dereceleri tamamen aynıdır** ($\text{rank}=2$). 

Ancak, dolanıklık entropisi hesaplandığında muazzam bir fark ortaya çıkar: İlk durum olan meşhur EPR çiftinin dolanıklık entropisi tam olarak **1** çıkarken; ikinci durumun dolanıklık entropisi ikili entropi fonksiyonu (binary entropy function) cinsinden **$H(\epsilon)$** değerine eşit olur. 

$$\lim_{\epsilon \to 0} H(\epsilon) = 0$$

olduğundan, $\epsilon$ sıfıra yaklaştıkça ikinci durumun dolanıklığı tamamen yok olmaya başlar. İşte tam da bu matematiksel nedenden dolayı, EPR çiftine iki kübitlik durumlar arasında tanımlanabilecek **"en yoğun/maksimum dolanık" (maximally entangled)** durum adını veririz; çünkü onun dolanıklık entropisi olası tüm iki-qubitlik kuantum durumları arasında ulaşılabilecek en yüksek (maksimum) sınırdadır.

## 2.3 Sır Paylaşımı (Secret sharing)

Kuantum dolanıklığı ve yoğunluk matrisleri hakkında şimdiye kadar öğrendiğimiz soyut kavramların çok önemli bir kriptografik uygulaması olan **sır paylaşımını** ele alacağız. 

Konuyu somutlaştırmak adına şu senaryoyu hayal edelim: Bir ülkenin nükleer silahları olduğunu ve bu silahların ateşlenebilmesi için hem kraliçenin (Alice) hem de kralın (Bob) mutlaka bir araya gelerek sistemi aktif etmesi gerektiğini varsayalım. Akla gelen ilk ilkel çözüm; nükleer fırlatma kodunun ($s = (s_1, \dots, s_n) \in \{0,1\}^n$) tam olarak yarısını Alice'e, diğer yarısını ise Bob'a vermektir. Böylece silahların aktif edilebilmesi için iki tarafın da kendi elindeki bilgi parçasını ifşa etmesi gerekecektir.

Ancak bu basit yöntemin çok büyük bir güvenlik açığı (zafiyeti) vardır: İki taraf da tek başınayken bile fırlatma kodunun tam olarak yarısı hakkında kesin bilgiye sahiptir. Peki, ya sadece tek bir bitlik ($n=1$) bir sırrımız olsaydı ne yapacaktık? (Kodu ortadan ikiye bölmek tek bir bit için pek de güvenli ve mantıklı bir çözüm olmazdı...)

Bir sır paylaşımı şemasının asıl hedefi; $s$ gizli bilgisini taraflar arasında öyle akıllıca parçalara (paylara/shares) bölmektir ki, yetkisiz veya tek başına hareket eden hiçbir taraf (bu örnekte tek başına Alice veya tek başına Bob) sır hakkında **en ufak bir bilgi dahi ediremesin**.

Daha önceki haftalarda öğrendiğimiz tek kullanımlık şerit (One-Time Pad) mantığını hatırlayacak olursak, klasik dünyada bundan çok daha iyi bir şema kurgulayabiliriz: Tamamen rastgele bir $r \in \{0,1\}^n$ bit dizisi seçelim ve bu $r$ dizesini doğrudan Alice'e verelim. Bob'u ise $r \oplus s$ şifreli dizesiyle ödüllendirelim. 

Bu harika senaryoda, ne tek başına Alice ne de tek başına Bob bireysel olarak orijinal $s$ sırrı hakkında hiçbir bilgiye sahip olamaz; çünkü iki tarafın elindeki paylar da dışarıdan tamamen homojen ve rastgele görünür. Ancak iki dürüst taraf bir araya gelip paylarını XOR işlemine tabi tuttuklarında ($(r) \oplus (r \oplus s) = s$), orijinal sırrı saniyeler içinde hatasızca geri elde edebilirler.

Yukarıdaki klasik örnekten görebileceğimiz üzere, rastgele klasik bitler yardımıyla tek bitlik bir $s$ sırrını Alice ve Bob arasında güvenle paylaştırabiliyoruz. Ancak, eğer bu iki tarafa klasik bitler yerine kuantum durumları (kübit) verirsek **çok daha gelişmiş ve güçlü kriptografik protokoller** inşa edebiliriz!

Alice ve Bob'a, kuantum mekaniğinin ve dolanıklık teorisinin temeli olan aşağıdaki **4 Bell Durumundan (Bell States)** rastgele seçilen bir tanesinin verildiği kuantum senaryosunu ele alalım :

$$\begin{aligned}

|\Phi^+\rangle_{AB} &= \frac{1}{\sqrt{2}}\Big( |00\rangle_{AB} + |11\rangle_{AB} \Big) \quad \text{--- (2.18)} & 
|\Phi^-\rangle_{AB} &= \frac{1}{\sqrt{2}}\Big( |00\rangle_{AB} - |11\rangle_{AB} \Big) \quad \text{--- (2.18)} \\
|\Psi^+\rangle_{AB} &= \frac{1}{\sqrt{2}}\Big( |01\rangle_{AB} + |10\rangle_{AB} \Big) \quad \text{--- (2.19)} &
|\Psi^-\rangle_{AB} &= \frac{1}{\sqrt{2}}\Big( |01\rangle_{AB} - |10\rangle_{AB} \Big) \quad \text{--- (2.19)}
\end{aligned}$$

Bu durumlara **Bell durumları** adı verilir. Dikkat ederseniz bu durumlar birbirine ortonormaldir ve bu sayede $\mathbb{C}^2 \otimes \mathbb{C}^2$ iki-kübitlik uzay için eksiksiz bir taban (baz) oluştururlar. 

Bölüm 01 de bu durumlardan biri olan EPR çifti ($|\Phi^+\rangle_{AB}$) için Alice'in sistemindeki indirgenmiş yoğunluk matrisini zaten hesaplamıştık. 

$$\begin{aligned}
\rho_A &= \text{tr}_B\big(|\Phi^+\rangle\langle\Phi^+|_{AB}\big) \\
&= \frac{1}{2} |0\rangle\langle0|_A \text{tr}_B\big(|0\rangle\langle0|_B\big) + \frac{1}{2} |0\rangle\langle1|_A \text{tr}_B\big(|0\rangle\langle1|_B\big) \\
&\quad + \frac{1}{2} |1\rangle\langle0|_A \text{tr}_B\big(|1\rangle\langle0|_B\big) + \frac{1}{2} |1\rangle\langle1|_A \text{tr}_B\big(|1\rangle\langle1|_B\big) \\
&= \frac{1}{2}\Big( |0\rangle\langle0|_A + |1\rangle\langle1|_A \Big) = \frac{\mathbb{I}_A}{2}
\end{aligned}$$


Her 4 Bell durumu için de Alice ($A$) veya Bob ($B$) üzerindeki indirgenmiş marjinal durumları hesaplamak bize her zaman kuruşu kuruşuna tamamen aynı sonucu verir:

$$\rho_{\Phi^+}^A = \rho_{\Phi^-}^A = \rho_{\Psi^+}^A = \rho_{\Psi^-}^A = \frac{\mathbb{I}_A}{2} \quad \text{--- (2.20)}$$

$$\rho_{\Phi^+}^B = \rho_{\Phi^-}^B = \rho_{\Psi^+}^B = \rho_{\Psi^-}^B = \frac{\mathbb{I}_B}{2} \quad \text{--- (2.21)}$$

Bu eşitliklerin fiziksel ve felsefi anlamı şudur: Her bir alt sistemdeki indirgenmiş durum tam bir **maksimum karma durumdur**. Dolayısıyla, Alice veya Bob tek başınayken, ellerindeki tekil kübit üzerinden hangi Bell durumuna sahip olduklarına dair **kesinlikle hiçbir bilgi edinemezler**. Ancak, Alice ve Bob bir araya geldiklerinde birleşik sistem üzerinde gerçekleştirecekleri ortak bir Bell tabanı ölçümü (Bell basis measurement) ile, eldeki durumun hangisi olduğunu %100 kesinlikle ayırt edebilirler. Bu ayırt edilebilirlik, kuantum sır paylaşımının anahtarıdır.

> 🟦 **Alıştırma 2.3.1 ** Şimdi sisteme üçüncü bir tarafın, yani Charlie'nin (başbakan) de dahil olduğunu varsayalım. Üç parçalı (tripartite) dolanık bir durum kullanarak şu üç farklı sır paylaşımı şemasını kurgulayınız:
> 1. Tek başına hiçbir taraf sırrı öğrenemesin, ancak üçü bir araya geldiğinde sırrı tamamen çözebilsinler.
> 2. Tek başına hiçbir taraf bilgi edinemesin, dahası **herhangi iki taraf bir araya gelse bile** sır hakkında hiçbir şey öğrenemesin; yalnızca üçü birden bir araya geldiğinde sırrı çözebilsinler.
> 3. Tek başına hiçbir taraf bilgi edinebilir olmasın, ancak **herhangi iki taraf bir araya geldiğinde** sırrı çözebilsinler.

**Çözümler ve Protokol Tasarımları:**

**1 & 2. Senaryoların Çözümü (Maksimum Korumalı [3,3] Şeması):**
Bu görevi başarmak için Alice, Bob ve Charlie arasında meşhur üç-kübitlik **GHZ durumunu** paylaştıralım:
$$|\text{GHZ}\rangle_{ABC} = \frac{1}{\sqrt{2}}\Big( |000\rangle_{ABC} + |111\rangle_{ABC} \Big)$$
*   **Gizlilik Kanıtı:** Herhangi tek bir tarafın marjinal matrisine ($\text{tr}_{BC}$) veya herhangi iki tarafın ortak marjinal matrisine ($\text{tr}_C$) baksak da indirgenmiş durumlar her zaman tamamen köşegen ve karmaşık kuantum fazlarından yoksundur ($\mathbb{I}/2$ veya $\mathbb{I}/4$ karışımıdır). Dolayısıyla, tek bir kişi ya da rastgele iki kişi bir araya gelse dahi sır hakkında **%0 bilgi** edinir.
*   **Çözüm Adımı:** Yalnızca üçü bir araya geldiğinde, birleşik durum uzayında ortak bir GHZ ölçümü ya da Hadamard/XOR adımları uygulayarak sırrı kusursuzca geri elde ederler.

**3. Senaryonun Çözümü (Eşikli [2,3] Şeması):**
Tek bir kişinin hiçbir şey bilmemesini ancak herhangi iki kişinin sırrı çözebilmesini istiyorsak, 3-kübitlik **W durumunu** veya permütasyon simetrili dolanık durumları kullanabiliriz:
$$|\text{W}\rangle_{ABC} = \frac{1}{\sqrt{3}}\Big( |100\rangle + |011\rangle + |101\rangle \Big)$$
Bu yapıda, herhangi bir parçayı yok ettiğimizde (kısmi iz aldığımızda), geriye kalan iki parçanın birleşik yoğunluk matrisi içinde hala güçlü kuantum dolanıklığı ve korelasyonu kalır. Böylece, Charlie olmasa bile Alice ve Bob kendi aralarında yapacakları bir ölçümle sırrı anında tetikleyebilirler.


```python
import numpy as np

# Temel tekli qubit durumları
q0 = np.array([1.0, 0.0])
q1 = np.array([0.0, 1.0])

# GHZ durumunun 8x1 boyutundaki ket vektörünü kuralım: (|000> + |111>) / sqrt(2)
v000 = np.kron(q0, np.kron(q0, q0))
v111 = np.kron(q1, np.kron(q1, q1))
ghz_ket = (v000 + v111) / np.sqrt(2)

# Birleşik 8x8 boyutundaki yoğunluk matrisi
rho_ABC = np.outer(ghz_ket, ghz_ket)

# Charlie'nin (3. qubit) uzayını yok edip Alice ve Bob'un 4x4'lük matrisini bulalım (Kısmi İz)
rho_AB_reduced = np.zeros((4, 4), dtype=complex)
# Blok matris köşegen marjinal toplam adımları
for i in range(4):
    for j in range(4):
        # Charlie'nin 0 ve 1 olduğu indisleri topluyoruz
        rho_AB_reduced[i, j] = rho_ABC[i*2, j*2] + rho_ABC[i*2+1, j*2+1]

print("Alice ve Bob'un ortak marjinal matrisi rho_AB:\n", np.real(rho_AB_reduced))
print("\nMatrisin köşegen dışı (off-diagonal) kuantum süperpozisyon elemanları tamamen 0 mı?:")
print(np.isclose(rho_AB_reduced[0, 3], 0.0), "(True olmalı, yani iki kişi bir araya gelse de sırrı çözemez!)")
```

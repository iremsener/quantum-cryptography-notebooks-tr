

## 1.1 Olasılık Notasyonu (Probability notation)

 Klasik olasılık teorisinin standart notasyonlarını kısaca hatırlatmak ile başlamak istiyorum bu bölüme. Detaylı bir altyapı için olasılık teorisi üzerine yazılmış pek çok temel kaynağa başvurulabilir.

$\mathcal{X}$ alfabesinden (kümesinden) değerler alan ve alfabesinin boyutu $|\mathcal{X}| = n$ olan kesikli (discrete) bir $X$ rastgele değişkenini ele alalım. $X$'in dağılımını $P_X(\cdot)$ ile, alfabesinin boyutunu ise $|X|$ ile göstereceğiz. $P_X(x)$ notasyonu, rastgele değişkenin belirli bir $x \in \mathcal{X}$ sembolünü alma olasılığını ifade eder. 

Dağılımın hangi değişkene ait olduğu bağlamdan netçe anlaşıldığında, şu pratik kısaltmaları (shorthands) kullanırız:
$$p_x = p(x) = P(X = x) = P_X(x)$$

Bir olasılık dağılımının geçerli olabilmesi için negatif olmayan değerlerden oluşması gerektiğini unutmamalıyız; yani her $x \in \mathcal{X}$ için $P_X(x) \geq 0$'dır. Ayrıca, $X$ normalize edilmelidir; bu da tüm olası sonuçların olasılıklarının toplamının tam olarak 1'e eşit olması gerektiği anlamına gelir:
$$\sum_{x \in \mathcal{X}} P_X(x) = 1$$


> 📝 **Örnek 1.1.1 ** $\mathcal{X} = \{1,2,3,4\}$ kümesinin 4 yüzlü adil bir piramit zarın yüzlerine karşılık geldiğini varsayalım. Eğer zar adilse, yani tüm yüzlerin gelme olasılığı birbirine eşitse, her $x \in \mathcal{X}$ için olasılık dağılımı şu şekilde olur:
> 
> $$P_X(x) = \frac{1}{4}$$
> 
> Yukarıdaki pratik kısaltmaları kullanırsak bu ifade $p_x = p(x) = 1/4 = 0.25$ olarak okunur. Bu durumda alfabenin boyutu (olası sonuçların sayısı) ise $|X| = 4$ olarak verilir.


Bir $X$ rastgele değişkeni, başka bir $Y$ rastgele değişkeni ile ilişkili (korele) olabilir. Bu durum, onların bir birleşik dağılıma (joint distribution) $P_{XY}(x,y)$ sahip olduğu anlamına gelir. Bu birleşik dağılım genel olarak bağımsız iki dağılımın doğrudan çarpımına eşit olmak zorunda değildir, yani:
$$P_{XY}(x,y) \neq P_X(x)P_Y(y)$$

Bu durum bizi **koşullu olasılık (conditional probability)** $P_{X|Y}(x|y)$ kavramına ulaştırır. Burada $P_{X|Y}(x|y)$ ifadesi, $Y$ değişkeninin $y$ değerini aldığı bilindiğinde (bu olaya koşullu olarak), $X$ değişkeninin $x$ değerini alma olasılığıdır. 

Hangi rastgele değişkenden bahsettiğimiz net olduğunda yine şu pratik kısaltmayı kullanırız:
$$p_{x|y} = p(x|y) = P(X = x \mid Y = y) = P_{X|Y}(x|y) \quad \text{--- (1.1)}$$



```python
import numpy as np

# 4 yüzlü adil piramit zar için olasılık dizisi (Her bir yüzün olasılığı 1/4)
olasiliklar = np.array([1/4] * 4)

# Normalizasyon şartını kontrol edelim: Toplam 1 mi?
toplam_olasilik = np.sum(olasiliklar)
print(f"4 Yüzlü Zar Olasılıkları Toplamı: {toplam_olasilik}")

# Simülasyon: Bu yeni zarı test etmek için 5 kez atalım
zar_sonuclari = np.random.choice(range(1, 5), size=5, p=olasiliklar)
print(f"Rastgele 5 Zar Atışı Sonucu: {zar_sonuclari}")
```

> 📝 **Örnek 1.1.2** Yukarıdaki adil 4 yüzlü zarı ve her zaman "4" gelen hileli (adil olmayan) başka bir zarı ele alalım. Yani hileli zar için alfabemiz yine $\mathcal{X} = \{1,2,3,4\}$ şeklindedir ancak olasılık dağılımı $P_X(4) = 1$ ve $x \neq 4$ olan durumlar için $P_X(x) = 0$ olur. 
> 
> $Y$ değişkeni, bu iki zardan (adil veya hileli) hangisini seçeceğimizi temsil etsin. Bu seçimi eşit olasılıkla yaptığımızı varsayalım: $\mathcal{Y} = \{\text{"adil"}, \text{"hileli"}\}$ ve $P_Y(\text{adil}) = 1/2$, $P_Y(\text{hileli}) = 1/2$ olsun. 
> 
> Bu durumda koşullu olasılıklarımız şu şekilde tanımlanır:
> *   $P_{X|Y}(x \mid \text{adil}) = 1/4$ (her $x$ için)
> *   $P_{X|Y}(4 \mid \text{hileli}) = 1$ ve $x \neq 4$ için $P_{X|Y}(x \mid \text{hileli}) = 0$


> 🟦 **Alıştırma 1.1.1** Adil veya hileli zar seçimi örneği için birleşik olasılık dağılımını ($P_{XY}(x,y)$) hesaplayınız.

**Çözüm:**
Birleşik olasılık formülü $P_{XY}(x,y) = P_Y(y) \cdot P_{X|Y}(x|y)$ şeklindedir. Seçtiğimiz zara göre tüm olasılıkları çarpalım:

1. **Adil Zar Seçildiğinde ($Y = \text{adil}$):**
   *   $P_{XY}(1, \text{adil}) = \frac{1}{2} \cdot \frac{1}{4} = \frac{1}{8}$
   *   $P_{XY}(2, \text{adil}) = \frac{1}{2} \cdot \frac{1}{4} = \frac{1}{8}$
   *   $P_{XY}(3, \text{adil}) = \frac{1}{2} \cdot \frac{1}{4} = \frac{1}{8}$
   *   $P_{XY}(4, \text{adil}) = \frac{1}{2} \cdot \frac{1}{4} = \frac{1}{8}$

2. **Hileli Zar Seçildiğinde ($Y = \text{hileli}$):**
   *   $P_{XY}(1, \text{hileli}) = \frac{1}{2} \cdot 0 = 0$
   *   $P_{XY}(2, \text{hileli}) = \frac{1}{2} \cdot 0 = 0$
   *   $P_{XY}(3, \text{hileli}) = \frac{1}{2} \cdot 0 = 0$
   *   $P_{XY}(4, \text{hileli}) = \frac{1}{2} \cdot 1 = \frac{1}{2}$ (yani $\frac{4}{8}$)


> 🟦 **Alıştırma 1.1.2** Yine adil veya hileli zarı $P_Y(\text{adil}) = P_Y(\text{hileli}) = 1/2$ olasılıklarıyla seçtiğimizi varsayalım ancak hangisini seçtiğimizi size söylemeyeyim. Bunun yerine, zar atışının sonucu olan $X$'i size göstereyim (Bende $Y$ var, sizde ise $X$ var). 
> 
> 1. Zarın sonucu **$X = 3$** geldiyse, en olası zar hangisidir? Yani $Y = \text{adil}$ olması mı yoksa $Y = \text{hileli}$ olması mı daha olasıdır?
> 2. Peki zarın sonucu **$X = 4$** gelseydi durum ne olurdu?

**Çözüm:**
Bu ters olasılık problemini çözmek için **Bayes Teoremini** ($P_{Y|X}(y|x) = \frac{P_{XY}(x,y)}{P_X(x)}$) kullanacağız. Öncelikle marjinal $P_X(x)$ olasılıklarını (yani zarın genel olarak o sayıyı getirme toplam olasılığını) bulmalıyız:
*   $P_X(3) = P_{XY}(3, \text{adil}) + P_{XY}(3, \text{hileli}) = \frac{1}{8} + 0 = \frac{1}{8}$
*   $P_X(4) = P_{XY}(4, \text{adil}) + P_{XY}(4, \text{hileli}) = \frac{1}{8} + \frac{1}{2} = \frac{5}{8}$

**1. Durum ($X = 3$ için):**
*   $P_{Y|X}(\text{adil} \mid 3) = \frac{P_{XY}(3, \text{adil})}{P_X(3)} = \frac{1/8}{1/8} = 1$ (%100)
*   $P_{Y|X}(\text{hileli} \mid 3) = \frac{P_{XY}(3, \text{hileli})}{P_X(3)} = \frac{0}{1/8} = 0$ (%0)
*   *Sonuç:* Eğer 3 geldiyse zar **kesinlikle adildir**, çünkü hileli zar asla 3 getiremez.

**2. Durum ($X = 4$ için):**
*   $P_{Y|X}(\text{adil} \mid 4) = \frac{P_{XY}(4, \text{adil})}{P_X(4)} = \frac{1/8}{5/8} = \frac{1}{5} = 0.20$ (%20)
*   $P_{Y|X}(\text{hileli} \mid 4) = \frac{P_{XY}(4, \text{hileli})}{P_X(4)} = \frac{4/8}{5/8} = \frac{4}{5} = 0.80$ (%80)
*   *Sonuç:* Eğer 4 geldiyse zarın **hileli olması çok daha olasıdır** (%80 ihtimalle hilelidir).


```python
# Birleşik olasılık tablosunu oluşturup Bayes hesabı yapalım
# Satırlar: X (1, 2, 3, 4) | Sütunlar: Y (0: adil, 1: hileli)
import numpy as np

P_XY = np.array([
    [1/8, 0],   # X=1
    [1/8, 0],   # X=2
    [1/8, 0],   # X=3
    [1/8, 1/2]  # X=4
])

# Marjinal P(X) hesaplama
P_X = np.sum(P_XY, axis=1)

# X=4 (indeks 3) geldiğinde Y'nin koşullu olasılıkları
P_Y_given_X4 = P_XY[3] / P_X[3]

print(f"X=4 geldiğinde zarın adil olma olasılığı: {P_Y_given_X4[0]:.2f}")
print(f"X=4 geldiğinde zarın hileli olma olasılığı: {P_Y_given_X4[1]:.2f}")
```

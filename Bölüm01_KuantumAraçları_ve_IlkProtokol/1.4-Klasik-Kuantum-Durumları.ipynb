### 1.4.1 Klasik Durumlar (Classical states)

Konuyu daha iyi anlamak adına ilk adım olarak, $X$ değişkeninin "klasik bilgi" olmasının tam olarak ne anlama geldiği üzerinde biraz duralım. Bu doğrultuda, klasik bit dizileri ($x$) üzerindeki bir olasılık dağılımını, yoğunluk matrisleri formalizmini kullanarak nasıl yeniden yazabileceğimizi incelemek ilginç bir yaklaşım sunar.

$\mathcal{X} = \{0, \dots, d-1\}$ alfabesindeki semboller üzerinde tanımlı, herhangi bir $x$ sembolünün gözlemlenme olasılığının $p_x$ ile verildiği klasik bir olasılık dağılımımız olduğunu varsayalım. Klasik bitleri (veya sayıları), kuantum durum uzayındaki standart taban elemanları ($\{|0\rangle, \dots, |d-1\rangle\}$) ile eşleştirerek; her bir olası $|x\rangle$ durumunu $p_x$ olasılığıyla hazırlayan bir kaynağı şu istatistiksel karışım (yoğunluk matrisi) ile ifade edebiliriz:

$$\rho = \sum_{x=0}^{d-1} p_x |x\rangle\langle x| \quad \text{--- (1.41)}$$


Buradaki $\rho$ matrisinin, köşegen (diagonal) elemanları doğrudan $p_x$ olasılıklarına eşit olan ve köşegen dışındaki tüm elemanları (off-diagonal) tamamen sıfır olan **köşegen bir yoğunluk matrisi** olduğuna dikkat edilmelidir. 

Bu yönüyle $\rho$, aslında klasik $p_x$ olasılık dağılımını kuantum dilinde ifade etmenin başka bir yolundan ibarettir. Gerçekten de, bu $\rho$ durumunu standart tabanda ölçtüğünüzde, tam olarak $p_x$ olasılığıyla "$x$" sonucunu elde edeceğinizi Born kuralı yardımıyla kolayca doğrulayabilirsiniz.


> 📝 **Örnek Alıştırma:** $\mathcal{X} = \{0, 1\}$ alfabesi üzerinde $p_0 = 0.7$ ve $p_1 = 0.3$ olasılıklarına sahip klasik bir dağılımı ele alalım.
> 1. Bu dağılımın kuantum dilindeki $\rho$ yoğunluk matrisini inşa ediniz.
> 2. Bu $\rho$ durumunu standart tabanda ölçtüğümüzde "0" görme olasılığının gerçekten $0.7$ çıktığını Born kuralı ($\langle 0 | \rho | 0 \rangle$) ile ispatlayınız.
> 
> **Çözüm:**
> 1. Formül (1.41) gereği dış çarpımları yazalım:
> 
> $$\rho = 0.7 |0\rangle\langle 0| + 0.3 |1\rangle\langle 1| = 0.7 \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} + 0.3 \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 0.7 & 0 \\ 0 & 0.3 \end{pmatrix}$$
> 
> 2. "0" sonucunu elde etme olasılığını hesaplayalım:
> 
> $$p(\text{"0"}) = \langle 0 | \rho | 0 \rangle = \begin{pmatrix} 1 & 0 \end{pmatrix} \begin{pmatrix} 0.7 & 0 \\ 0 & 0.3 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 1 & 0 \end{pmatrix} \begin{pmatrix} 0.7 \\ 0 \end{pmatrix} = 0.7$$
> 
> İspat tamamlanmıştır. Klasik olasılık, kuantum matris formunda da tamamen korunmaktadır.


```python
import numpy as np

# Klasik olasılık değerleri
p0 = 0.7
p1 = 0.3

# Formül 1.41'e göre köşegen kuantum matrisini inşa edelim
rho_klasik = np.array([[p0, 0.0],
                       [0.0, p1]])

print("Klasik olasılığın kuantum matris karşılığı:\n", rho_klasik)

# Born kuralı ile ölçüm simülasyonu: |0> durumu üzerinden izdüşüm
qubit_0 = np.array([1.0, 0.0])
olasilik_0 = np.dot(qubit_0, np.dot(rho_klasik, qubit_0))

print(f"\nMatris üzerinden ölçülen '0' gelme olasılığı: {olasilik_0:.1f}")
print(f"Klasik p0 değeriyle eşleşiyor mu?: {np.isclose(olasilik_0, p0)}")
```

> 🟦 **Tanım 1.4.1 — Klasik Durum (c-state).** Durum uzayı $\mathbb{C}^d$ olan bir $X$ sistemini ve bu uzay için standart tabanı $\{|x\rangle\}_{x=0}^{d-1}$ olarak ele alalım. Eğer ilgili $\rho_X$ yoğunluk matrisi, $X$ uzayının standart tabanında köşegen (diagonal) bir yapıya sahipse, bu $X$ sistemi bir **klasik durum (classical state / c-state)** içindedir. Yani $\rho$ matrisi şu formdadır:
> 
> $$\rho = \sum_{x=0}^{d-1} p_x |x\rangle\langle x| \quad \text{--- (1.42)}$$
> 
> Burada $\{p_x\}_{x=0}^{d-1}$ herhangi bir normalize edilmiş olasılık dağılımıdır.


Kuantum şifrelemede, genellikle kısmen klasik ve kısmen kuantum olan birleşik durumlarla karşılaşırız. Bu yapıyı anlamak için Alice ve Bob (veya bir saldırgan) arasında paylaşılan şu özgün senaryoyu ele alalım:

> 📝 **Örnek 1.4.1 ** Bir kuantum protokolünde, $1/2$ olasılıkla Alice için $|0\rangle\langle0|_X$ klasik bitini hazırlarken Bob'a $\rho_0^Q = |0\rangle\langle0|$ saf kuantum durumunu verdiğimizi varsayalım. Yine $1/2$ olasılıkla Alice için $|1\rangle\langle1|_X$ bitini hazırlarken Bob'a $\rho_1^Q = |-\rangle\langle-|$ durumunu iletelim. 
> 
> Bu iki sistemin paylaştığı birleşik durum, kuantum literatüründe bir **klasik-kuantum durumu (cq-state)** olarak adlandırılır ve şu formülü alır:
> 
> $$\rho_{XQ} = \frac{1}{2} |0\rangle\langle0|_X \otimes |0\rangle\langle0| + \frac{1}{2} |1\rangle\langle1|_X \otimes |-\rangle\langle-| = \frac{1}{2} \sum_{x \in \{0,1\}} |x\rangle\langle x|_X \otimes \rho_x^Q \quad \text{--- (1.43)}$$

Bu senaryoda dikkat edilmesi gereken felsefi nokta şudur: Alice kendi elindeki klasik $x$ bitini kesin olarak bilmektedir. Ancak ileride göreceğimiz üzere, Bob'a verilen kuantum durumları ($\rho_0^Q$ ve $\rho_1^Q$) birbirine tamamen dik (orthogonal) olmadığı, aksine aralarında bir **üst üste binme (overlap)** olduğu için Bob, Alice'in elindeki $x$ değerini %100 kesinlikle asla ayırt edemez. Kuantum anahtar dağıtımının (QKD) temel güvenliği tam olarak bu ayırt edilemezliğe dayanır.


> 🟦 **Tanım 1.4.2 — Klasik-Kuantum Durumu (cq-state).** Genel olarak bir klasik-kuantum durumu (veya cq-durumu) şu matematiksel forma sahiptir:
> 
> $$\rho_{XQ} = \sum_{x} p_x |x\rangle\langle x|_X \otimes \rho_x^Q \quad \text{--- (1.44)}$$
> 
> Yani bu sistem, klasik bir saklayıcı (register) olan $X$ sistemi ile kuantum özellikler barındıran bir $Q$ saklayıcısının birleşiminden oluşur. Eğer ortamda kuantum parçası ($Q$) yoksa, durum doğrudan Tanım 1.4.1'deki klasik duruma indirgenir.

Kuantum kriptografisi uygulamalarında, $x$ harfi genellikle dürüst tarafların (Alice) protokol sırasında ürettiği (ve kısmen gizli olan) klasik bit dizilerini temsil eder. $\rho_x^Q$ ise, bir saldırganın (Eve) bu klasik dizgeyle korele (ilişkili) olacak şekilde sistemden gizlice topladığı kuantum bilgilerini ifade eder. Kuantum bilgi literatüründe klasik saklayıcıları göstermek için geleneksel olarak $X, Y, Z$ harfleri tercih edilirken, kuantum sistemleri için diğer harfler (örn. $Q, A, B$) rezerve edilmiştir.


```python
import numpy as np

# Standart tekli kübit durumları
q0 = np.array([1, 0])
q1 = np.array([0, 1])
minus = (q0 - q1) / np.sqrt(2)

# Alice'in klasik register taban matrisleri (|0><0| ve |1><1|)
X0 = np.outer(q0, q0)
X1 = np.outer(q1, q1)

# Bob'un (veya saldırganın) özgün kuantum durumları (rho_0^Q ve rho_1^Q)
rho_0_Q = np.outer(q0, q0)       # |0><0|
rho_1_Q = np.outer(minus, minus)   # |-><-|

# Formül 1.43'e göre cq-durum matrisini (Kronecker çarpımı ile) inşa edelim
# Olasılıklar yarı yarıya (0.5)
rho_XQ = 0.5 * np.kron(X0, rho_0_Q) + 0.5 * np.kron(X1, rho_1_Q)

print("İnşa edilen 4x4 boyutundaki cq-durum matrisi:\n", rho_XQ)

# Blok matris yapısını kontrol etmek için köşegen toplamı (Trace) doğrulaması
print(f"\nBileşke matrisin izi (tr): {np.trace(rho_XQ)} (1.0 olmalı)")
```

## 1.5  Ölçümler 

Kuantum mekaniği, hem kuantum bilgi teorisinde hem de şifrelemede hayati bir rol oynayan çok daha gelişmiş ve esnek bir ölçüm kavramına izin verir. 

Bir yandan, kuantum bilgi teorisinde birden fazla kuantum durumunu birbirinden ayırt etme (state discrimination) gibi belirli görevler, bu genelleştirilmiş ölçümler kullanılarak çok daha yüksek bir verimlilikle çözülebilir. Diğer yandan, güvenlik ve casus analizi açısından bakıldığında, kuantum kriptografisinde bir saldırganın (Eve) yapabileceği tüm olası fiziksel müdahaleleri ve en genel saldırı türlerini kapsayacak şekilde güvenliği kanıtlamak istiyorsak, bu en genel ölçüm mekanizmasını kullanmamız gerekiyor.


### 1.5.1 POVM'lar (POVMs)

Eğer ölçümden sonra kuantum sisteminin durumuna ne olduğuyla (çökme sonrası yeni haliyle) ilgilenmiyor, **sadece hangi ölçüm sonucunun hangi olasılıkla elde edileceğiyle** ilgileniyorsak; kuantum mekaniğinde izin verilen en genel ölçüm türü **Pozitif Operatör Değerli Ölçüm (Positive Operator-Valued Measure)**, yani kısaca **POVM** formalizmi ile tanımlanır:

> 🟦 **Tanım 1.5.1 — POVM.** $\mathbb{C}^d$ uzayında tanımlı bir POVM, aşağıdaki normalizasyon (tamamlayıcılık) şartını sağlayan bir pozitif yarı-tanımlı operatörler kümesidir $\{M_x\}_{x \in \mathcal{X}}$:
> 
> $$\sum_{x} M_x = \mathbb{I}_{\mathbb{C}^d} \quad \text{--- (1.45)}$$
> 
> Buradaki $x$ alt indisi, ölçüm sonucunu etiketlemek için kullanılır. Rastgele bir $\rho$ yoğunluk matrisi üzerinde bu ölçüm yapıldığında, $x$ sonucunu gözlemleme olasılığı $p_x$, **Born Kuralı**'nın genel formu kullanılarak şu iz (trace) işlemiyle hesaplanır:
> 
> $$p_x = \text{tr}(M_x \rho) \quad \text{--- (1.46)}$$


> 📝 **Örnek 1.5.1 ** Standart taban ölçümlerinin aslında bu genel POVM tanımının çok özel bir alt durumu olduğunu **Hadamard tabanı** ($\{|+\rangle, |-\rangle\}$) üzerinden gösterelim. 
> 
> Bir $|\psi\rangle = \alpha_+ |+\rangle + \alpha_- |-\rangle$ kuantum durumunu ele alalım. Bu durumu Hadamard tabanında ölçtüğümüzde $+$ sonucunu elde etme olasılığımızın $|\alpha_+|^2$ olduğunu zaten biliyoruz. Şimdi bunu POVM diline aktaralım. Her bir $x \in \{+, -\}$ sonucu için POVM operatörlerimizi birer izdüşüm operatörü (projector) olarak tanımlayalım:
> 
> $$M_+ = |+\rangle\langle+| \quad \text{ve} \quad M_- = |-\rangle\langle-|$$
> 
> Bu operatörler pozitif yarı-tanımlıdır ve kareleri kendilerine eşittir ($M^2 = M$). Ayrıca taban özelliklerinden dolayı toplamları birim matrisi verir: $\sum_x M_x = |+\rangle\langle+| + |-\rangle\langle-| = \mathbb{I}$. Dolayısıyla bu geçerli bir POVM setidir. Şimdi Born kuralını (1.46) kullanarak olasılığı adım adım hesaplayalım ($\rho = |\psi\rangle\langle\psi|$ olmak üzere):
> 
> $$p_+ = \text{tr}(M_+ \rho) = \text{tr}(|+\rangle\langle+| \psi \rangle\langle\psi|)$$
> 
> İz işleminin döngüsel (cyclic) özelliğini kullanarak $\langle+| \psi \rangle$ karmaşık sayısını başa alalım:
> 
> $$p_+ = \langle+|\psi\rangle \text{tr}(|+\rangle\langle\psi|) = \langle+|\psi\rangle \langle\psi|+\rangle = |\langle+|\psi\rangle|^2$$
> 
> Durum vektörümüzün açık halini yerine koyarsak ($\langle+|+\rangle = 1$ ve $\langle+|-\rangle = 0$ diklik şartlarıyla):
> 
> $$p_+ = |\langle+| (\alpha_+ |+\rangle + \alpha_- |-\rangle) |^2 = |\alpha_+ \cdot 1 + \alpha_- \cdot 0|^2 = |\alpha_+|^2$$
> 
> Görüldüğü üzere genelleştirilmiş POVM formülü, bildiğimiz standart taban ölçüm olasılıklarını kusursuz bir şekilde doğrulamaktadır.


```python
import numpy as np

# Temel Kübit durumları ve Hadamard durumları
q0 = np.array([1, 0])
q1 = np.array([0, 1])
plus = (q0 + q1) / np.sqrt(2)
minus = (q0 - q1) / np.sqrt(2)

# Test için rastgele katsayılarla bir |psi> durumu hazırlayalım (Örn: alpha_+ = 0.8, alpha_- = 0.6)
# Not: 0.8^2 + 0.6^2 = 0.64 + 0.36 = 1.0 (Normalize)
alpha_plus = 0.8
alpha_minus = 0.6
psi_ket = (alpha_plus * plus) + (alpha_minus * minus)
rho = np.outer(psi_ket, psi_ket) # Saf durum yoğunluk matrisi

# POVM Operatörlerimizi tanımlayalım (M_+ ve M_- izdüşüm matrisleri)
M_plus = np.outer(plus, plus)
M_minus = np.outer(minus, minus)

# POVM şartının kontrolü: Toplamları Birim Matris (I) veriyor mu?
I_kontrol = M_plus + M_minus
print("POVM Matris Toplamı:\n", I_kontrol)

# Born Kuralı ile Olasılık Hesabı: p_+ = tr(M_+ * rho)
p_plus_teorik = alpha_plus ** 2
p_plus_povm = np.trace(np.dot(M_plus, rho))

print(f"\nTeorik Beklenen Olasılık (|alpha_+|^2): {p_plus_teorik:.2f}")
print(f"POVM Formülü ile Ölçülen Olasılık: {p_plus_povm:.2f}")
print(f"Sonuçlar uyuşuyor mu?: {np.isclose(p_plus_povm, p_plus_teorik)}")
```

## 1.6 Kısmi İz (The partial trace)

Çoklu kübit içeren birleşik bir kuantum durumu verildiğinde, bu kübitlerin sadece bir alt kümesiyle (parçasıyla) ilişkili olan "kısmi durumu" (partial state) nasıl yalnız bırakıp yazabiliriz? 

Daha genel bir ifadeyle; $\rho_{AB}$ matrisinin $\mathbb{C}^{d_A} \otimes \mathbb{C}^{d_B}$ tensör çarpım uzayında tanımlı bir birleşik yoğunluk matrisi olduğunu varsayalım. Bu $\rho_{AB}$ durumunun $A$ sistemine karşılık gelen kısmını Alice, $B$ sistemine karşılık gelen kısmını ise Bob elinde tutuyor olsun. Alice'in sisteminin tek başına sahip olduğu $\rho_A$ durumunu nasıl tarif edebiliriz?


### 1.6.1 İşlemsel Bir Bakış Açısı (An operational viewpoint)

Bizi birleşik $\rho_{AB}$ durumundan Alice'in tekil $\rho_A$ durumuna götüren matematikselleme işlemine **kısmi iz (partial trace)** adı verilir. Bu işleme birazdan tamamen soyut ve matematiksel bir tanım vereceğiz. Ancak öncesinde, probleme daha sezgisel ve işlemsel bir bakış açısıyla yaklaşmaya çalışalım. 

İlk olarak en kolay senaryoyu ele alalım: Eğer birleşik durum bir çarpım durumu (product state) ise; yani $\rho_A$ ve $\rho_B$ geçerli birer yoğunluk matrisi olmak üzere $\rho_{AB} = \rho_A \otimes \rho_B$ şeklinde veriliyorsa, Alice'in sisteminin durumunun doğrudan $\rho_A$ olduğu son derece açıktır. 

Bir adım daha ileri giderek biraz daha karmaşık bir senaryoyu düşünelim: Birleşik durumumuz, çarpım durumlarının olasılıksal bir karışımı olsun (ilerleyen bölümlerde bu tür durumlara **ayrılabilir durumlar / separable states** diyeceğiz):
$$\rho_{AB} = \sum_{i} p_i \rho_i^A \otimes \rho_i^B \quad \text{--- (1.49)}$$

Bu senaryoda, Alice'in sisteminin $p_i$ olasılığıyla $\rho_i^A$ durumunda olduğunu söylemek son derece doğal bir yaklaşım olacaktır. Dolayısıyla Alice'in nihai marjinal durumu ağırlıklı toplamla şu şekilde yazılır: $\rho_A = \sum_{i} p_i \rho_i^A$.


Peki, ya elimizde hiçbir çarpım yapısına uymayan, tamamen genel ve dolanık (entangled) bir $\rho_{AB}$ matrisi varsa ne yapacağız? 

Buradaki temel fikir, Bob'un kendi sistemi ($B$) üzerinde rastgele bir $\{|u_x\rangle\}$ tabanını kullanarak eksiksiz bir ölçüm gerçekleştirdiğini hayal etmektir. Bu ölçüm işlemini modellemek adına, Alice ve Bob'un birleşik sistemi üzerinde tanımlı bir POVM seti kurgulayalım. Ölçüm sırasında Alice kendi sistemi üzerinde hiçbir işlem yapmadığı için, onun tarafındaki operatörü birim matris ($\mathbb{I}_A$) olarak bırakabiliriz:

$$M_x = \mathbb{I}_A \otimes |u_x\rangle\langle u_x|_B$$

Bu ifadenin geçerli ve normalize bir POVM seti tanımladığını kendiniz de kolayca kontrol edebilirsiniz ($\sum M_x = \mathbb{I}_A \otimes \mathbb{I}_B = \mathbb{I}$). Dahası, bu bir izdüşüm ölçümü (projective measurement) olduğu için Kraus operatörlerini de doğrudan POVM operatörlerine eşit seçebiliriz: $A_x = M_x$. 

Bu tanımlar doğrultusunda, Bob ölçüm yapıp $x$ sonucunu elde ettiğinde, birleşik sistemin **ölçüm sonrası durumu (post-measurement state)** genel kuantum kuralları gereği şu kesirli formülü alır:

$$\rho_{|x}^{AB} = \frac{M_x \rho_{AB} M_x}{\text{tr}(M_x \rho_{AB})} = \frac{\Big( (\mathbb{I}_A \otimes |u_x\rangle\langle u_x|_B) \, \rho_{AB} \, (\mathbb{I}_A \otimes |u_x\rangle\langle u_x|_B) \Big)}{\text{tr}\Big( (\mathbb{I}_A \otimes |u_x\rangle\langle u_x|_B) \, \rho_{AB} \Big)} \quad \text{--- (1.50)}$$


```python
import numpy as np

# Denemek için 4x4 boyutunda rastgele bir birleşik rho_AB matrisi oluşturalım
# (Doğrulama amaçlı rastgele pozitif yarı-tanımlı matris)
A = np.random.rand(4, 4) + 1j *np.random.rand(4, 4)
rho_AB = np.dot(A, A.conj().T)
rho_AB /= np.trace(rho_AB) # Normalizasyon: tr(rho) = 1

# Bob'un |0> tabanında ölçüm yaptığını varsayalım (u_x = |0>)
q0 = np.array([1, 0])
proj_B = np.outer(q0, q0) # |0><0| matrisi (2x2)

# Alice hiçbir şey yapmıyor (I_A = 2x2 birim matris)
I_A = np.eye(2)

# Birleşik POVM elemanı M_0 = I_A tensor proj_B
M_0 = np.kron(I_A, proj_B)

# Pay kısmındaki çarpım: M_0 * rho_AB * M_0
pay_matris = np.linalg.multi_dot([M_0, rho_AB, M_0])

print("Bob |0> sonucunu gözlemlediğinde oluşan (pay) matris yapısı:\n", np.round(pay_matris, 3))
print(f"\nMatrisin yeni boyutu: {pay_matris.shape} (Hala birleşik uzayda)")
```

Bob'un yaptığı bu ölçüm sonucunda $x$ çıktısını elde etme olasılığı $q_x = \text{tr}\big( (\mathbb{I} \otimes |u_x\rangle\langle u_x|) \rho_{AB} \big)$ ile verilir. Toplam olasılık teorisi gereği, Alice'in nihai durumunu ($\rho_A$) bulmak için her bir $x$ sonucu sonrasında oluşan durumları kendi olasılık ağırlıklarıyla ($q_x$) çarparak toplamamız gerekir. 

Önceki  (1.50) numaralı ifadenin payda kısmındaki olasılık terimi ile dışarıdaki $q_x$ çarpanı birbirini kusursuz bir şekilde götürür (sadeleşir). Sonuç olarak, Alice'in nihai marjinal durumunu veren **kısmi iz** formülü şu bağımsız forma kavuşur:

$$\rho_A = \sum_{x} q_x \frac{\Big( (\mathbb{I} \otimes \langle u_x|) \rho_{AB} (\mathbb{I} \otimes |u_x\rangle) \Big)_A}{\text{tr}\big( (\mathbb{I} \otimes |u_x\rangle\langle u_x|) \rho_{AB} \big)} = \sum_{x} (\mathbb{I} \otimes \langle u_x|) \rho_{AB} (\mathbb{I} \otimes |u_x\rangle) \quad \text{--- (1.51)}$$


Yukarıdaki ifadeyi mantıklı argümanlar kullanarak türetmiş olsak da, üzerinde durulması ve endişelenilmesi gereken çok önemli bir detay mevcuttur: Alice'in bulduğumuz bu nihai $\rho_A$ durumu, Bob'un ölçüm yaparken tamamen rastgele seçtiği $\{|u_x\rangle\}$ tabanına (bazına) bağımlı mıdır? 

Cevap kesinlikle **hayır, bağımlı olmamalıdır**. Çünkü tüm kuantum bilgi teorisi, Alice'in lokal sistemindeki indirgenmiş durumun, Bob'un kendi tarafında yapacağı hiçbir lokal işlem veya taban seçimiyle uzaktan değiştirilemeyeceği felsefesine dayanır (bu durum görelilik teorisiyle uyumlu olan **No-Signaling / Işık Hızından Hızlı İletişim Kurulamaz** ilkesinin de temelidir). 

> ⚠️ **Çok Kritik Bir İnce Nokta:** Bu bağımsızlık kuralı, yalnızca Alice'in Bob'un elde ettiği ölçüm sonuçlarından **haberdar olmadığı (sonuçları öğrenmediği)** sürece geçerlidir! Eğer Bob ölçüm sonucunu Alice'e söyler ve Alice belirli bir $x$ sonucuna göre durumunu koşullandırırsa, o zaman hikaye tamamen değişir ve indirgenmiş durum tabana göre farklılık gösterir. Kuantum kriptografisindeki casus tespit mekanizmaları bu ince nüansa dayanır.


> 🟦 **Alıştırma 1.6.1 ** Denklem (1.51) ile tanımlanan $\rho_A$ durumunun, Bob'un ölçüm için seçtiği $\{|u_x\rangle\}$ tabanından tamamen bağımsız olduğunu matematiksel olarak ispatlayınız.

**İspat:**
İspatı gerçekleştirmek için alıştırmadaki ipuçlarını adım adım takip edelim.

**Adım 1:** Herhangi iki yoğunluk matrisi $\rho$ ve $\sigma$, tüm birim $|\phi\rangle$ vektörleri için $\langle\phi|\rho|\phi\rangle = \langle\phi|\sigma|\phi\rangle$ şartını sağlıyorsa, bu iki matris matematiksel olarak birbirine birebir eşittir ($\rho = \sigma$). O halde, rastgele bir birim $|\phi\rangle_A$ vektörü seçelim ve Alice'in durumunun bu vektör üzerindeki izdüşüm beklentisini ($\langle\phi|\rho_A|\phi\rangle$) hesaplayalım.

**Adım 2:** Denklem (1.51)'i bu sandviç çarpımında yerine koyalım:
$$\langle\phi|\rho_A|\phi\rangle = \langle\phi| \left( \sum_{x} (\mathbb{I} \otimes \langle u_x|) \rho_{AB} (\mathbb{I} \otimes |u_x\rangle) \right) |\phi\rangle$$

Doğrusallık (linearity) özelliğini kullanarak toplam sembolünü ve vektörleri içeri dağıtalım:
$$\langle\phi|\rho_A|\phi\rangle = \sum_{x} \Big( (\langle\phi|_A \otimes \langle u_x|_B) \, \rho_{AB} \, (|\phi\rangle_A \otimes |u_x\rangle_B) \Big)$$

**Adım 3:** İz (trace) işleminin döngüsel (cyclic) özelliğini ve dış çarpım ilişkisini ($\langle v| M |v\rangle = \text{tr}(M |v\rangle\langle v|)$) kullanalım. Ortadaki birleşik vektörlerin dış çarpımını matrisin yanına getirelim:
$$\langle\phi|\rho_A|\phi\rangle = \sum_{x} \text{tr} \Big( \rho_{AB} \cdot \big( (|\phi\rangle\langle\phi|_A) \otimes (|u_x\rangle\langle u_x|_B) \big) \Big)$$

İz işleminin doğrusallığı sayesinde toplam sembolünü matrislerin içine kaydırabiliriz:
$$\langle\phi|\rho_A|\phi\rangle = \text{tr} \left( \rho_{AB} \cdot \left[ (|\phi\rangle\langle\phi|_A) \otimes \left( \sum_{x} |u_x\rangle\langle u_x|_B \right) \right] \right)$$

**Adım 4:** $\{|u_x\rangle\}$ kümesi Bob'un uzayı için eksiksiz bir ortonormal taban olduğundan, **Tamamlayıcılık (Olasılık Korunumu / Closure Relation)** şartı gereği bu toplam bize doğrudan Bob'un uzayındaki birim matrisi verir: $\sum_{x} |u_x\rangle\langle u_x|_B = \mathbb{I}_B$. Bu ifadeyi yerine koyarsak:
$$\langle\phi|\rho_A|\phi\rangle = \text{tr} \Big( \rho_{AB} \cdot \big( (|\phi\rangle\langle\phi|_A) \otimes \mathbb{I}_B \big) \Big)$$

**Sonuç:** Elde edilen en son eşitlik incelendiğinde, ifadenin içinde Bob'un seçtiği $\{|u_x\rangle\}$ tabanına dair hiçbir terim kalmamıştır. Denklem yalnızca birleşik $\rho_{AB}$ matrisine ve Alice'in test vektörüne bağlıdır. Dolayısıyla, Alice'in indirgenmiş kuantum durumu $\rho_A$, **Bob'un lokal taban seçiminden tamamen bağımsızdır**. 


> 📝 **Örnek 1.6.1 — EPR Çifti Üzerinden Kısmi İz Hesaplaması.** Dolanık kuantum durumlarının en meşhur örneği olan EPR çiftini ele alalım:
> 
> $$|\text{EPR}\rangle_{AB} = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle) \quad \text{--- (1.51)}$$
> 
> Bu saf durumu, dış çarpım (outer product) yöntemiyle birleşik bir yoğunluk operatörü (matrisi) olarak yazarsak şu ifadeyi elde ederiz:
> 
> $$\rho_{AB} = |\text{EPR}\rangle\langle\text{EPR}|_{AB} = \frac{1}{2}\Big( |00\rangle\langle00| + |00\rangle\langle11| + |11\rangle\langle00| + |11\rangle\langle11| \Big) \quad \text{--- (1.52)}$$


Şimdi Bob'un kendi sistemini ($B$) standart hesapsal tabanda ölçtüğünü varsayalım. Alice'in sistemini ($A$) de denkleme dahil ederek, bu ölçüm işlemini şu iki elemanlı POVM setiyle modelleyebiliriz:
*   $M_0 = \mathbb{I}_A \otimes |0\rangle\langle0|_B$
*   $M_1 = \mathbb{I}_A \otimes |1\rangle\langle1|_B$

Bob'un ölçüm sonucunda "0" çıktısını elde etme olasılığı ($q_0$) Born kuralı yardımıyla şu şekilde hesaplanır:

$$q_0 = \text{tr}(M_0 \rho_{AB}) = \frac{1}{2} \text{tr}\Big( (\mathbb{I} \otimes |0\rangle\langle0|)(|00\rangle\langle00| + |00\rangle\langle11| + |11\rangle\langle00| + |11\rangle\langle11|) \Big)$$

Matris çarpımlarını ve diklik şartlarını ($\langle0|1\rangle = 0$, $\langle0|0\rangle = 1$) göz önünde bulundurarak iz değerini açarsak:

$$q_0 = \frac{1}{2}(1 + 0 + 0 + 0) = \frac{1}{2}$$

Benzer şekilde Bob'un "1" sonucunu elde etme olasılığı da $q_1 = 1/2$ olarak bulunur. Bu olasılıkları kullanarak Alice'in elinde kalan nihai ölçüm sonrası durumu ($\rho_A$) hesaplarsak:

$$\rho_A = \frac{1}{2}(\mathbb{I} \otimes \langle0|)\rho_{AB}(\mathbb{I} \otimes |0\rangle) + \frac{1}{2}(\mathbb{I} \otimes \langle1|)\rho_{AB}(\mathbb{I} \otimes |1\rangle) = \frac{1}{2}|0\rangle\langle0| + \frac{1}{2}|1\rangle\langle1| = \frac{\mathbb{I}_A}{2}$$

Görüldüğü üzere, bileşke sistem tamamen saf bir EPR durumunda olmasına rağmen, alt sistemlerden birini yalnız bıraktığımızda elimizde hiçbir kuantum korelasyonu kalmayan, tamamen belirsiz bir **maksimum karma durum (maximally mixed state)** kalır.


> 🟦 **Alıştırma:** Bob'un ölçüm işlemini standart taban yerine **Hadamard tabanında** ($\{|+\rangle, |-\rangle\}$) gerçekleştirdiğini varsayarak aynı kısmi iz hesaplamasını yeniden yapınız. Alice'in elinde kalan nihai durumun yine tamamen aynı ($\mathbb{I}_A/2$) çıktığını ispatlayınız.

**İspat:**
Hafta 0 (Week 0) notlarında, EPR çiftinin Hadamard tabanındaki katsayılarının standart taban ile birebir aynı kaldığını ispatlamıştık:
$$|\text{EPR}\rangle_{AB} = \frac{1}{\sqrt{2}}(|++\rangle + |--\rangle)$$

Bu durumu yoğunluk matrisi olarak açalım:
$$\rho_{AB} = \frac{1}{2}\Big( |++\rangle\langle++| + |++\rangle\langle--| + |--\rangle\langle++| + |--\rangle\langle--| \Big)$$

Bob'un Hadamard tabanındaki ölçümünü modellemek adına yeni POVM operatörlerimizi tanımlayalım:
*   $M_+ = \mathbb{I}_A \otimes |+\rangle\langle+|_B$
*   $M_- = \mathbb{I}_A \otimes |-\rangle\langle-|_B$

1. **Olasılıkların Hesaplanması ($q_+$ ve $q_-$):**
   Hadamard taban elemanlarının kendi aralarındaki diklik şartlarını ($\langle+|-\rangle = 0$, $\langle+|+\rangle = 1$) kullanarak Born kuralını işletelim:
   $$q_+ = \text{tr}(M_+ \rho_{AB}) = \frac{1}{2}(1 + 0 + 0 + 0) = \frac{1}{2}$$
   Benzer şekilde $q_- = 1/2$ elde edilir.

2. **Alice'in İndirgenmiş Durumunun Hesaplanması ($\rho_A$):**
   Kısmi iz formülünde değerleri yerine koyalım:
   $$\rho_A = q_+ (\mathbb{I} \otimes \langle+|)\rho_{AB}(\mathbb{I} \otimes |+\rangle) + q_- (\mathbb{I} \otimes \langle-|)\rho_{AB}(\mathbb{I} \otimes |-\rangle)$$
   $$\rho_A = \frac{1}{2}|+\rangle\langle+| + \frac{1}{2}|-\rangle\langle-|$$

Hadamard taban elemanlarının dış çarpım toplamının da bize birim matrisi verdiğini (tamamlayıcılık ilişkisi / closure relation) biliyoruz: $|+\rangle\langle+| + |-\rangle\langle-| = \mathbb{I}_A$. Formülde yerine yazarsak:
$$\rho_A = \frac{\mathbb{I}_A}{2}$$

**Sonuç:** Bob kendi lokal sisteminde hangi tabanı seçerse seçsin (ister standart taban ister Hadamard tabanı), Alice'in indirgenmiş marjinal durumu kuruşu kuruşuna tamamen aynı ($\mathbb{I}/2$) kalmaktadır. Bu durum bir önceki sayfada teorik olarak ispatladığımız taban bağımsızlığı (No-Signaling) kuralını pratik bir örnek üzerinden kusursuzca doğrular.


```python
import numpy as np

# Temel kuantum durumları (|0> ve |1> ket vektörleri)
q0 = np.array([1, 0])
q1 = np.array([0, 1])

# |EPR> = (|00> + |11>) / sqrt(2) durumunun 4x1 boyutundaki ket vektörü
# np.kron([1,0], [1,0]) = [1,0,0,0] yani |00> durumu
# np.kron([0,1], [0,1]) = [0,0,0,1] yani |11> durumu
epr_ket = (np.kron(q0, q0) + np.kron(q1, q1)) / np.sqrt(2)

# Birleşik 4x4 yoğunluk matrisi rho_AB
rho_AB = np.outer(epr_ket, epr_ket)

# Kısmi İz Fonksiyonu (Bob'un uzayını yok edip Alice'in 2x2 matrisini bırakan pratik blok iz)
def partial_trace_B(matrix_4x4):
    # 4x4 matrisin blok elemanlarından Bob'un sistemini marjinalleştirme
    rho_A = np.zeros((2, 2), dtype=complex)
    
    # Alice |0><0| bloğu: Sol üst bloğun izi (0,0 ve 1,1 elemanları)
    rho_A[0, 0] = matrix_4x4[0, 0] + matrix_4x4[1, 1]
    
    # Alice |0><1| bloğu: Sağ üst bloğun izi (0,2 ve 1,3 elemanları)
    rho_A[0, 1] = matrix_4x4[0, 2] + matrix_4x4[1, 3]
    
    # Alice |1><0| bloğu: Sol alt bloğun izi (2,0 ve 3,1 elemanları)
    rho_A[1, 0] = matrix_4x4[2, 0] + matrix_4x4[3, 1]
    
    # Alice |1><1| bloğu: Sağ alt bloğun izi (2,2 ve 3,3 elemanları)
    rho_A[1, 1] = matrix_4x4[2, 2] + matrix_4x4[3, 3]
    
    return rho_A

# Alice'in marjinal durumunu bulalım
rho_A_final = partial_trace_B(rho_AB)

print("Alice'in indirgenmiş yoğunluk matrisi (rho_A):\n", np.real(rho_A_final))
print("\nTeorik maksimum karma durum matrisi (I/2):\n", np.eye(2) / 2)
print(f"\nİki matris birbirine eşit mi?: {np.allclose(rho_A_final, np.eye(2)/2)}")
```

### 1.6.2 Matematiksel Bir Tanım (A mathematical definition)

Kısmi izin ne olması gerektiğine dair yukarıda geliştirdiğimiz "işlemsel" yaklaşımdan güç alarak, şimdi bu işlemin kesin ve soyut matematiksel tanımını ortaya koyabiliriz.

> 🟦 **Tanım 1.6.1 — Kısmi İz (Partial Trace).** $\mathbb{C}^{d_A} \otimes \mathbb{C}^{d_B}$ tensör çarpım uzayında tanımlı, genel bir $\rho_{AB}$ birleşik kuantum durumunu ele alalım. Bu durum, indis katsayıları $\gamma_{ij}^{kl}$ olmak üzere şu şekilde ifade edilebilir:
> 
> $$\rho_{AB} = \sum_{ijkl} \gamma_{ij}^{kl} |i\rangle\langle j|_A \otimes |k\rangle\langle l|_B \quad \text{--- (1.53)}$$
> 
> Burada $\{|i\rangle_A, |j\rangle_A\}$ ve $\{|k\rangle_B, |l\rangle_B\}$ kümeleri sırasıyla $A$ ve $B$ sistemlerine ait bağımsız ortonormal tabanları temsil eder. Bu durumda, **B sistemi üzerinden alınan kısmi iz** ($\text{tr}_B$) matematiksel olarak şu formülle tanımlanır:
> 
> $$\rho_A = \text{tr}_B(\rho_{AB}) = \sum_{ijkl} \gamma_{ij}^{kl} |i\rangle\langle j|_A \otimes \text{tr}(|k\rangle\langle l|_B) = \sum_{ij} \left( \sum_{k} \gamma_{ij}^{kk} \right) |i\rangle\langle j|_A \quad \text{--- (1.54)}$$


Tamamen benzer bir mantıkla, eğer Alice'in sistemini yok sayıp sadece Bob'un elindeki kuantum durumunu yalnız bırakmak istiyorsak, **A sistemi üzerinden kısmi iz** ($\text{tr}_A$) alma işlemini şu şekilde uygulayabiliriz:
> 
> $$\rho_B = \text{tr}_A(\rho_{AB}) = \sum_{ijkl} \gamma_{ij}^{kl} \text{tr}(|i\rangle\langle j|_A) \otimes |k\rangle\langle l|_B = \sum_{kl} \left( \sum_{j} \gamma_{jj}^{kl} \right) |k\rangle\langle l|_B \quad \text{--- (1.55)}$$

Elde edilen bu $\rho_A$ ve $\rho_B$ marjinal kuantum durumları, kuantum literatüründe yaygın olarak **indirgenmiş durumlar (reduced states)** adıyla da anılır.


💡 **Denklemleri Okuma ve Uygulama Kolaylığı:**
Yukarıdaki (1.54) ve (1.55) numaralı formüller ilk bakışta yoğun görünse de, matris dünyasındaki karşılığı son derece pratiktir:
1. **$\text{tr}_B$ (Bob'u yok etme):** Birleşik $4 \times 4$ boyutundaki matrisin içinde, Bob'un indislerinin aynı olduğu ($k=l$, yani köşegen) elemanları kendi aralarında toplarız. Sonuçta Alice'e ait $2 \times 2$ boyutunda bir matris kalır.
2. **$\text{tr}_A$ (Alice'i yok etme):** Bu kez matris içindeki Alice'in indislerinin aynı olduğu ($i=j$) elemanları kendi aralarında toplarız. Sonuçta Bob'a ait $2 \times 2$ boyutunda bir matris kalır.


```python
import numpy as np

# 4x4 boyutunda örnek bir birleşik durum matrisi oluşturalım (Örn: EPR çifti matrisi)
q0 = np.array([1, 0])
q1 = np.array([0, 1])
epr_ket = (np.kron(q0, q0) + np.kron(q1, q1)) / np.sqrt(2)
rho_AB = np.outer(epr_ket, epr_ket)

# Formül 1.53'teki gibi matrisi 4 boyutlu bir tensöre (i, j, k, l indislerine) dönüştürelim
# Her bir eksen 2 boyutludur: (Alice_i, Alice_j, Bob_k, Bob_l) -> Toplam 2x2x2x2 = 16 eleman
rho_tensor = rho_AB.reshape(2, 2, 2, 2)

# Formül 1.54: Bob üzerinden kısmi iz (k ve l eksenlerinin köşegenini alıp topluyoruz)
# axis1=2 (k indisi), axis2=3 (l indisi) üzerinden iz alma işlemi
rho_A_reduced = np.trace(rho_tensor, axis1=2, axis2=3)

# Formül 1.55: Alice üzerinden kısmi iz (i ve j eksenlerinin köşegenini alıp topluyoruz)
# axis1=0 (i indisi), axis2=1 (j indisi) üzerinden iz alma işlemi
rho_B_reduced = np.trace(rho_tensor, axis1=0, axis2=1)

print("Matematiksel tanıma göre hesaplanan İndirgenmiş Durum rho_A:\n", rho_A_reduced)
print("\nMatematiksel tanıma göre hesaplanan İndirgenmiş Durum rho_B:\n", rho_B_reduced)
```

> 📝 **Örnek 1.6.2 — Matematiksel Tanım İle EPR Kısmi İz Hesabı.** Birleşik sistemimizin yine meşhur EPR çifti durumu olduğunu varsayalım:
> 
> $$|\text{EPR}\rangle_{AB} = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$$
> 
> Bu duruma karşılık gelen $\rho_{AB} = |\text{EPR}\rangle\langle\text{EPR}|_{AB}$ yoğunluk matrisini ele alalım. Bir önceki bölümde Tanım 1.6.1'de verdiğimiz soyut doğrusal kuralı doğrudan bu matrisin terimlerine uygulayarak Bob üzerinden kısmi iz ($\text{tr}_B$) hesabını şu şekilde dağıtabiliriz:
> 
> $$\text{tr}_B(\rho_{AB}) = \frac{1}{2}\Big( |0\rangle\langle0| \otimes \text{tr}(|0\rangle\langle0|) + |0\rangle\langle1| \otimes \text{tr}(|0\rangle\langle1|) + |1\rangle\langle0| \otimes \text{tr}(|1\rangle\langle0|) + |1\rangle\langle1| \otimes \text{tr}(|1\rangle\langle1|) \Big) \quad \text{--- (1.56)}$$


İz işleminin döngüsel özelliği ve iç çarpım kuralları gereği, zıt taban elemanlarının iz değerlerinin tamamen sıfıra eşit olduğunu biliyoruz: $\text{tr}(|0\rangle\langle1|) = \langle1|0\rangle = 0$ ve benzer şekilde $\text{tr}(|1\rangle\langle0|) = \langle0|1\rangle = 0$. 

Buna karşılık, aynı taban elemanlarının iz değerleri ise doğrudan 1'e eşittir: $\text{tr}(|0\rangle\langle0|) = \langle0|0\rangle = 1$ ve $\text{tr}(|1\rangle\langle1|) = \langle1|1\rangle = 1$. Bu değerleri (1.56) numaralı denklemde yerlerine koyarsak, ortadaki zıt terimler tamamen elenir ve Alice'in indirgenmiş durumu şu şekilde kalır:

$$\text{tr}_B(\rho_{AB}) = \frac{1}{2}\Big( |0\rangle\langle0| \cdot 1 + 0 + 0 + |1\rangle\langle1| \cdot 1 \Big) = \frac{1}{2}\Big( |0\rangle\langle0| + |1\rangle\langle1| \Big) = \frac{\mathbb{I}_A}{2} \quad \text{--- (1.57)}$$

Benzer şekilde, Alice'in sistemini yok sayıp sadece Bob'un qubitinin indirgenmiş durumuna bakmak adına A sistemi üzerinden kısmi iz ($\text{tr}_A$) işlemi aldığımızda da sonucun kusursuz bir simetriyle tamamen aynı çıktığını kendiniz de kolayca görebilirsiniz:

$$\text{tr}_A(\rho_{AB}) = \frac{\mathbb{I}_B}{2} \quad \text{--- (1.58)}$$


> 🟦 **Alıştırma 1.2.2** Eğer birleşik kuantum durumumuz $\rho_{AB} = |\Phi\rangle\langle\Phi|$ şeklinde bir **Singlet (Tekil)** durum ise, yani durum şu şekilde veriliyorsa:
> 
> $$|\Phi\rangle = \frac{1}{\sqrt{2}}(|01\rangle - |10\rangle)$$
> 
> Bu sistem için Alice ve Bob'un indirgenmiş $\rho_A$ ve $\rho_B$ durumlarını ayrı ayrı hesaplayınız.

**Çözüm:**
1. Öncelikle bu saf durumun dış çarpım yöntemiyle birleşik yoğunluk matrisini oluşturalım:
   $$\rho_{AB} = |\Phi\rangle\langle\Phi| = \frac{1}{2}\Big( |01\rangle\langle01| - |01\rangle\langle10| - |10\rangle\langle01| + |10\rangle\langle10| \Big)$$

2. **Alice'in İndirgenmiş Durumu ($\rho_A = \text{tr}_B(\rho_{AB})$) İçin Hesaplama:**
   Tanım 1.6.1'deki doğrusal iz dağıtma kuralını işletelim:
   $$\rho_A = \frac{1}{2}\Big( |0\rangle\langle0| \otimes \text{tr}(|1\rangle\langle1|) - |0\rangle\langle1| \otimes \text{tr}(|1\rangle\langle0|) - |1\rangle\langle0| \otimes \text{tr}(|0\rangle\langle1|) + |1\rangle\langle1| \otimes \text{tr}(|0\rangle\langle0|) \Big)$$
   
   İç çarpım kurallarını uygulayalım ($\text{tr}(|1\rangle\langle1|)=1$, $\text{tr}(|1\rangle\langle0|)=0$, $\text{tr}(|0\rangle\langle1|)=0$, $\text{tr}(|0\rangle\langle0|)=1$):
   $$\rho_A = \frac{1}{2}\Big( |0\rangle\langle0| \cdot 1 - 0 - 0 + |1\rangle\langle1| \cdot 1 \Big) = \frac{1}{2}\Big( |0\rangle\langle0| + |1\rangle\langle1| \Big) = \frac{\mathbb{I}_A}{2}$$

3. **Bob'un İndirgenmiş Durumu ($\rho_B = \text{tr}_A(\rho_{AB})$) İçin Hesaplama:**
   Bu kez iz işlemini Alice'in uzayına uygulayalım:
   $$\rho_B = \frac{1}{2}\Big( \text{tr}(|0\rangle\langle0|) \otimes |1\rangle\langle1| - \text{tr}(|0\rangle\langle1|) \otimes |1\rangle\langle0| - \text{tr}(|1\rangle\langle0|) \otimes |0\rangle\langle1| + \text{tr}(|1\rangle\langle1|) \otimes |0\rangle\langle0| \Big)$$
   $$\rho_B = \frac{1}{2}\Big( 1 \cdot |1\rangle\langle1| - 0 - 0 + 1 \cdot |0\rangle\langle0| \Big) = \frac{1}{2}\Big( |0\rangle\langle0| + |1\rangle\langle1| \Big) = \frac{\mathbb{I}_B}{2}$$

**Sonuç:** Singlet (Tekil) kuantum durumunda da kısmi iz aldığımızda, aradaki eksi işaretine rağmen, hem Alice hem de Bob için indirgenmiş kuantum durumları yine tamamen birer **maksimum karma durum ($\mathbb{I}/2$)** olarak elde edilir.

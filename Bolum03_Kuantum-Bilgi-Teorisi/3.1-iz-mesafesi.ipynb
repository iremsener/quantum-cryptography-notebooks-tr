# İz Mesafesi (Trace Distance)

# İki Kuantum Durumunun Ayırt Edilebilirliği: Operasyonel ve Geometrik Yaklaşım

Kuantum mekaniğinde önceden bilinmeyen bir kuantum sisteminin hangi durumda olduğunu tek bir ölçümle kesin olarak belirlemek, durumlar ortogonal (dik) değilse imkansızdır. Bu durum, bizi **operasyonel ayırt edilebilirlik** kavramına götürür.

## 1. Operasyonel Anlam ve Helstrom Sınırı

Kuantum bilgi teorisinde ayırt edilebilirlik problemi genellikle bir oyun gibi kurgulanır: Bize $p_1$ olasılığıyla $\rho$ durumu, $p_2$ olasılığıyla ise $\sigma$ durumu veriliyor ($p_1 + p_2 = 1$). Amacımız, yapacağımız tek bir kuantum ölçümü ile sistemin aslında hangi durumda olduğunu maksimum başarı olasılığıyla tahmin etmektir.

### 1.1. POVM (Positive Operator-Valued Measure) ile Ölçüm Modeli

Kuantum mekaniğindeki en genel ölçüm türü POVM'dir. Bu iki durumlu problemi çözmek için iki ölçüm operatörü tanımlarız:
*   $E_1$: "Sistem $\rho$ durumundadır" sonucunu veren ölçüm operatörü.
*   $E_2$: "Sistem $\sigma$ durumundadır" sonucunu veren ölçüm operatörü.

Fiziksel olarak geçerli bir ölçüm yapabilmemiz için bu operatörlerin toplamının birim matrisi vermesi ($E_1 + E_2 = I$) ve pozitif yarı-tanımlı ($E_1, E_2 \ge 0$) olmaları gerekir. Seçtiğimiz herhangi bir ölçüm düzeneği için ortalama başarı olasılığımız şudur:

$$ P_{succ} = p_1 \text{Tr}(E_1 \rho) + p_2 \text{Tr}(E_2 \sigma) $$

### 1.2. Genelleştirilmiş Helstrom Sınırı

Carl Helstrom, doğanın bize izin verdiği *en iyi (optimum)* ölçüm operatörlerini ($E_1$ ve $E_2$) seçtiğimizde, bu başarı olasılığının ulaşabileceği mutlak tavanı matematiksel olarak kanıtlamıştır. Bu üst sınıra **Helstrom Sınırı** denir:

$$ P_{succ} = \frac{1}{2} + \frac{1}{2} || p_1 \rho - p_2 \sigma ||_1 $$

### 1.3. Eşit Olasılıklar ve İz Mesafesi

Eğer durumların bize verilme olasılıkları eşitse (yani kimse hile yapmıyorsa, $p_1 = p_2 = 1/2$), Helstrom formülü doğrudan **İz Mesafesi** ($T$) tanımına indirgenir:

$$ P_{succ} = \frac{1}{2} + \frac{1}{2} T(\rho, \sigma) $$

Burada $T(\rho, \sigma)$, iki durum arasındaki matrisel uzaklıktır ve $0 \le T \le 1$ aralığında değer alır:
*   **Mükemmel Ayırt Edilebilirlik ($T = 1$):** Durumlar birbirine tamamen ortogonaldir. Optimum ölçümle onları %100 başarıyla ($P_{succ} = 1$) ayırt edebiliriz (Örneğin saf $|0\rangle$ ve $|1\rangle$ durumları).
*   **Tamamen Belirsizlik ($T = 0$):** Durumlar birebir aynıdır ($\rho = \sigma$). Elimizdeki en mükemmel kuantum sensörü bile bu durumda sadece yazı-tura atarak rastgele tahmin yapabilir ($P_{succ} = 0.5$).

## 2. Geometrik Anlam ve İz Mesafesi (Trace Distance)

İz mesafesi, durum uzayındaki iki matris arasındaki metrik uzaklığı ifade eder ve matematiksel olarak şu şekilde tanımlanır:

$$ T(\rho, \sigma) = \frac{1}{2} ||\rho - \sigma||_1 = \frac{1}{2} \text{Tr} \left( \sqrt{(\rho - \sigma)^\dagger (\rho - \sigma)} \right) $$

**Tek Qubitlik Sistemler İçin Geometrik Yorum:**
Tek bir qubit için yoğunluk matrisleri Bloch küresi üzerinde bir vektör ile ifade edilebilir:
$$ \rho = \frac{1}{2}(I + \vec{r} \cdot \vec{\sigma}) \quad \text{ve} \quad \sigma = \frac{1}{2}(I + \vec{s} \cdot \vec{\sigma}) $$

Bu durumda iz mesafesi, Bloch küresinin içindeki bu iki vektör arasındaki Öklid mesafesinin tam yarısına eşittir:

$$ T(\rho, \sigma) = \frac{1}{2} |\vec{r} - \vec{s}| $$

Yani geometrik olarak iki durum Bloch küresinde ne kadar uzaksa, operasyonel olarak onları ayırt etmek o kadar kolaydır.

## 2.1 İz Mesafesi ve Öklid Mesafesi Arasındaki İlşkinin İspatı

Tek qubitlik sistemler için yoğunluk matrisleri, Pauli matrisleri ($\sigma_x, \sigma_y, \sigma_z$) tabanında şu şekilde yazılır:
$$ \rho = \frac{1}{2}(I + \vec{r} \cdot \vec{\sigma}) $$
$$ \sigma = \frac{1}{2}(I + \vec{s} \cdot \vec{\sigma}) $$

İki durum arasındaki fark matrisini ($\Delta$) tanımlayalım:
$$ \Delta = \rho - \sigma = \frac{1}{2} (\vec{r} - \vec{s}) \cdot \vec{\sigma} $$

Kolaylık sağlaması için $\vec{v} = \vec{r} - \vec{s}$ diyelim. Bu durumda fark matrisi $\Delta = \frac{1}{2} (\vec{v} \cdot \vec{\sigma})$ olur. Pauli matrislerinin lineer kombinasyonlarının özdeğerleri, kombinasyon vektörünün uzunluğuna bağlıdır. 

Yani, $\vec{v} \cdot \vec{\sigma}$ matrisinin özdeğerleri $\pm |\vec{v}|$ şeklindedir. Dolayısıyla, $\Delta$ matrisinin özdeğerleri ($\lambda_1, \lambda_2$) şunlardır:
$$ \lambda_1 = \frac{1}{2}|\vec{v}|, \quad \lambda_2 = -\frac{1}{2}|\vec{v}| $$

İz normu (Trace Norm) $||A||_1$, bir matrisin tekil değerlerinin (Hermityen matrisler için özdeğerlerinin mutlak değerlerinin) toplamıdır:
$$ ||\rho - \sigma||_1 = |\lambda_1| + |\lambda_2| = \frac{1}{2}|\vec{v}| + \frac{1}{2}|\vec{v}| = |\vec{v}| = |\vec{r} - \vec{s}| $$

İz mesafesi $T(\rho, \sigma)$, iz normunun yarısı olarak tanımlandığı için nihai sonucu elde ederiz:
$$ T(\rho, \sigma) = \frac{1}{2} ||\rho - \sigma||_1 = \frac{1}{2} |\vec{r} - \vec{s}| $$

**Fiziksel Yorum:**
*   **Saf Durumlar (Pure States):** Bloch küresinin yüzeyindedirler ($|\vec{r}| = 1$ ve $|\vec{s}| = 1$). İki durum birbirine tam ortogonal (dik) ise, küre üzerinde birbirlerine zıt kutuplarda yer alırlar. Aralarındaki mesafe $|\vec{r} - \vec{s}| = 2$ olur ve İz Mesafesi $T = 1$ çıkar (Mükemmel ayırt edilebilir).
*   **Karışık Durumlar (Mixed States):** Kürenin içindedirler. Merkeze yaklaştıkça sistem "gürültülü" veya "belirsiz" hale gelir. Merkezdeki tamamen karışık durum ($I/2$) için vektör uzunluğu sıfırdır ($\vec{r} = 0$).

```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit.quantum_info import DensityMatrix, Pauli
from qiskit.visualization import plot_bloch_vector
```

```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit.quantum_info import (
    Statevector,
    DensityMatrix,
    state_fidelity
)
from qiskit.visualization import plot_bloch_vector


# 1. Kuantum durumlarının tanımlanması

# |0> durumu
psi_1 = Statevector([1, 0])
rho = DensityMatrix(psi_1)

# |+> durumu
psi_2 = Statevector([
    1 / np.sqrt(2),
    1 / np.sqrt(2)
])
sigma = DensityMatrix(psi_2)


# 2. İz mesafesinin doğrudan hesaplanması
# D(rho, sigma) = 1/2 ||rho - sigma||_1

difference = rho.data - sigma.data

singular_values = np.linalg.svd(
    difference,
    compute_uv=False
)

t_dist = 0.5 * np.sum(singular_values)


# 3. Helstrom başarı olasılığı
# Eşit önsel olasılıklar için:
# P_success = 1/2 (1 + D)

p_succ = 0.5 * (1 + t_dist)


# 4. Fidelity

fidelity = state_fidelity(rho, sigma)


# Sonuçların yazdırılması

print("--- Kuantum Durumlarının Karşılaştırması ---")
print(f"Trace Distance : {t_dist:.4f}")
print(f"Maksimum Başarı Olasılığı: %{p_succ * 100:.2f}")
print(f"Fidelity        : {fidelity:.4f}")


# 5. Bloch vektörlerinin hesaplanması

X = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)

Y = np.array([
    [0, -1j],
    [1j, 0]
], dtype=complex)

Z = np.array([
    [1, 0],
    [0, -1]
], dtype=complex)

pauli_matrices = [X, Y, Z]


def get_bloch_vector(density_matrix):
    """Tek kübit yoğunluk matrisinin Bloch vektörünü hesaplar."""

    return np.array([
        np.real(np.trace(density_matrix.data @ pauli))
        for pauli in pauli_matrices
    ])


rho_bloch = get_bloch_vector(rho)
sigma_bloch = get_bloch_vector(sigma)

print(f"Rho Bloch vektörü   : {rho_bloch}")
print(f"Sigma Bloch vektörü : {sigma_bloch}")


# 6. Bloch kürelerinin yan yana çizilmesi

fig = plt.figure(figsize=(12, 5))

ax1 = fig.add_subplot(1, 2, 1, projection="3d")
ax2 = fig.add_subplot(1, 2, 2, projection="3d")

plot_bloch_vector(
    rho_bloch,
    title=r"$\rho = |0\rangle\langle0|$",
    ax=ax1
)

plot_bloch_vector(
    sigma_bloch,
    title=r"$\sigma = |+\rangle\langle+|$",
    ax=ax2
)

fig.suptitle(
    "Kuantum Durumlarının Bloch Küresinde Gösterimi",
    fontsize=15
)

plt.tight_layout()
plt.show()
```

## Sonuçların Kavramsal ve Fiziksel Analizi

Elde edilen sonuçlar ve Bloch küresi görselleştirmesi, kuantum durumlarının doğası hakkında bize üç temel evrensel kuralı göstermektedir:

### 1. Vektör Uzunluğu ve Sistemin Saflığı (Purity)
Bloch küresindeki okların (vektörlerin) kürenin dış yüzeyine dokunmaması ve iç bölgede kalması, sistemin "saf" (pure) bir durumda olmadığını kanıtlar. Kuantum sistem çevreyle etkileşime girdiğinde veya bir gürültüye maruz kaldığında bilgi kaybı yaşar ve "karışık" (mixed) bir duruma çöker. Vektör merkeze ne kadar yakınsa, sistemdeki belirsizlik ve entropi o kadar yüksektir.

### 2. Geometri ve Kuantum Cebirinin Kusursuz Uyumu
Farklı yöntemlerle (biri tamamen 3 boyutlu Öklid geometrisiyle görselleştirerek, diğeri soyut matris cebiri ve iz normu kullanarak) hesaplanan mesafelerin birebir aynı çıkması çok önemli bir ispat barındırır. Bu durum, tek qubitlik sistemlerde kuantum mekaniğinin soyut lineer cebir kurallarının, sezgisel ve görsel 3D geometriye kusursuz bir şekilde tercüme edilebildiğini gösterir.

### 3. Operasyonel Ayırt Edilebilirlik Neden Sınırlıdır?
Hesaplanan maksimum başarı olasılığının ne mutlak kesinlikte ne de tamamen rastgele çıkması, doğanın bize koyduğu temel bir sınırı temsil eder. 
Sistemin kapalı bir kutu içindeki iki durumdan birini kesin olarak tahmin edememesinin iki ana sebebi vardır:
*   **Ortogonallik Eksikliği:** Durumlar birbirine tam zıt (dik) değildir.
*   **Gürültü (Belirsizlik):** Durumların saf olmaması, yani bilgi kaybı içermesi.

Sonuç olarak; kuantum algoritmaları veya ölçüm cihazları (POVM) ne kadar kusursuz tasarlanırsa tasarlansın, doğa bu iki gürültülü durumu tek bir ölçümle kesin olarak ayırmamıza izin vermez. Bize yalnızca belirli bir güven aralığında, istatistiksel bir başarı ihtimali sunar.

## 3. İz Mesafesinin Temel Özellikleri (Matematiksel ve Fiziksel)

İz mesafesi, sadece bir metrik olmanın ötesinde, kuantum bilgi teorisinin temel yasalarını yansıtan fiziksel özelliklere sahiptir.

### 3.1. Üniter Değişmezlik (Unitary Invariance)
Kapalı bir kuantum sistemi dış çevreyle etkileşime girmeden üniter bir evrim ($U$) geçirirse, durumlar arasındaki ayırt edilebilirlik değişmez:
$$ T(U\rho U^\dagger, U\sigma U^\dagger) = T(\rho, \sigma) $$
**Fiziksel Anlamı:** Kuantum algoritmalarında (örneğin kuantum makine öğrenmesi modellerinde) uyguladığımız kapı dizileri (ansatz) sadece durumları uzayda döndürür, durumların birbirine olan göreceli mesafesini değiştirmez. Bilgi kaybolmaz.

### 3.2. CPTP Kanalları Altında Büzülme ve Kraus Operatörleri

Eğer sistem açık bir sistemse ve çevreyle etkileşime giriyorsa, bu süreç üniter bir matris ile değil, bir **CPTP (Completely Positive Trace-Preserving)** kanalı olan $\mathcal{E}$ ile modellenir. Bu kanal uygulandığında iz mesafesi (ayırt edilebilirlik) asla artamaz:
$$T(\mathcal{E}(\rho), \mathcal{E}(\sigma)) \le T(\rho, \sigma)$$

#### CPTP Ne Anlama Gelir?
Gerçek dünyadaki herhangi bir fiziksel sürecin geçerli olabilmesi için şu iki kurala uyması şarttır:
*   **Tamamen Pozitif (Completely Positive - CP):** Yoğunluk matrisi, durumların olasılıklarını barındırır ve olasılıklar negatif olamaz. Sistemin tamamı veya başka bir sistemle dolanık (entangled) bir parçası bu kanaldan geçse bile, sonuçtaki yoğunluk matrisinin özdeğerleri her zaman $\ge 0$ olmalıdır.
*   **İzi Koruyan (Trace-Preserving - TP):** Bir ölçüm yapılmadan önce, bir sistemin herhangi bir durumda bulunma olasılıklarının toplamı (matrisin izi) daima $1$'e eşit olmalıdır ($\text{Tr}(\rho) = 1$). Sistem dış ortama bilgi sızdırsa bile, evrendeki toplam olasılık $1$ olarak kalır.

#### Kraus Operatörleri ile Matematiksel Gösterim
Kuantum kanalının (gürültü, ölçüm, çevre ile etkileşim vb.) sisteme etkisi, **Kraus Operatörleri** ($K_k$) kullanılarak zarif bir şekilde ifade edilir. Sistemin kanaldan geçtikten sonraki yeni durumu şu toplam ile hesaplanır:
$$\mathcal{E}(\rho) = \sum_k K_k \rho K_k^\dagger$$

Bu işlemin olasılıkların toplamını (izi) koruması için, Kraus operatörlerinin şu tamlık bağıntısını sağlaması matematiksel bir zorunluluktur:
$$\sum_k K_k^\dagger K_k = I$$

**Büzülmenin Fiziksel Nedeni (Data Processing Inequality):**
Çevre (environment) ile etkileşim yaşandığında, sistemdeki bilginin bir kısmı çevreye sızar. Biz sadece ana sistemi gözlemleyebildiğimiz ve çevreye sızan bilgiyi kaybettiğimiz için, iki farklı başlangıç durumu ($\rho$ ve $\sigma$) birbirine benzemeye başlar. Bilgi teorisindeki "Veri İşleme Eşitsizliği" gereği, eksik veri üzerinde yapılan hiçbir yerel işlem iki durumu birbirinden daha kolay ayırt edilebilir hale getiremez.

### 3.3. Fuchs-van de Graaf Eşitsizlikleri
İz mesafesi ($T$), kuantum durumlarının benzerliğini ölçen **Fidelity (Sadakat - $F$)** ile doğrudan ilişkilidir. $F(\rho, \sigma) = \left( \text{Tr} \sqrt{\sqrt{\rho} \sigma \sqrt{\rho}} \right)^2$ olmak üzere, bu iki metrik şu eşitsizlikle birbirine bağlıdır:
$$ 1 - \sqrt{F(\rho, \sigma)} \le T(\rho, \sigma) \le \sqrt{1 - F(\rho, \sigma)} $$

```python
import numpy as np

from qiskit.quantum_info import (
    DensityMatrix,
    Operator,
    SuperOp
)
from qiskit.circuit.library import RXGate
from qiskit_aer.noise import depolarizing_error


# İz mesafesini kendimiz hesaplıyoruz
def calculate_trace_distance(state_1, state_2):
    """
    D(rho, sigma) = 1/2 ||rho - sigma||_1
    """

    difference = state_1.data - state_2.data

    singular_values = np.linalg.svd(
        difference,
        compute_uv=False
    )

    return 0.5 * np.sum(singular_values)


# 1. Birbirine yakın iki karışık durum

rho = DensityMatrix([
    [0.7, 0.2],
    [0.2, 0.3]
])

sigma = DensityMatrix([
    [0.6, 0.1],
    [0.1, 0.4]
])


# Başlangıç iz mesafesi

t_initial = calculate_trace_distance(rho, sigma)


# 2. Üniter evrim

rx_unitary = Operator(RXGate(np.pi / 4))

rho_unitary = rho.evolve(rx_unitary)
sigma_unitary = sigma.evolve(rx_unitary)

t_unitary = calculate_trace_distance(
    rho_unitary,
    sigma_unitary
)


# 3. Depolarize edici CPTP kanal

noise_error = depolarizing_error(
    param=0.3,
    num_qubits=1
)

# QuantumError nesnesini DensityMatrix.evolve() tarafından
# uygulanabilen bir SuperOp kanalına dönüştürüyoruz.
noise_channel = SuperOp(
    noise_error.to_instruction()
)

rho_noisy = rho.evolve(noise_channel)
sigma_noisy = sigma.evolve(noise_channel)

t_noisy = calculate_trace_distance(
    rho_noisy,
    sigma_noisy
)


# 4. Sonuçlar

print("--- İz Mesafesi Özellikleri ---")

print(
    f"Başlangıç İz Mesafesi            : "
    f"{t_initial:.5f}"
)

print(
    f"Üniter Evrim Sonrası T_unitary   : "
    f"{t_unitary:.5f}"
    "  -> Bilgi korunur, T değişmez"
)

print(
    f"Gürültü Sonrası T_noisy          : "
    f"{t_noisy:.5f}"
    "  -> Büzülme ilkesi, T azalır"
)
```

## 4. Büyük Resim: Kapalı ve Açık Sistemlerin Doğası Üzerine Genel Bir Yorum

İz mesafesinin üniter evrim altında sabit kalması ve gürültü altında büzülmesi, kuantum mekaniğinin temel yasalarının çok net bir özetidir. Sayılardan bağımsız olarak, bu matematiksel özellikler bize evrenin bilgiyi nasıl işlediğine dair şu üç evrensel kuralı söyler:

### 4.1. Bilginin Korunumu (Kapalı Sistemler)
İdeal bir kuantum bilgisayarında veya kapalı bir sistemde, uyguladığımız kuantum kapıları (üniter operasyonlar) durumu sadece durum uzayında (Bloch küresinde) "döndürür". 
Bu dönüşüm, iki farklı kuantum durumu arasındaki göreceli mesafeyi asla değiştirmez. Yani, eğer dışarıya bilgi sızmıyorsa (ölçüm veya gürültü yoksa), sistem başlangıçtaki "ayırt edilebilirliğini" sonsuza kadar korur. Kuantum evrimi, bilgiyi dönüştürür ama asla yok etmez.

### 4.2. Bilginin Silinmesi (Açık Sistemler ve Gürültü)
Kuantum sistemleri gerçek dünyada izole edilemez. Çevreyle olan en ufak bir etkileşim (dekoherans) bir kuantum kanalı (CPTP haritası) olarak işlev görür. 
Gürültünün evrensel doğası **büzülme (contractivity)** ilkesiyle tanımlanır. Gürültü, farklı kuantum durumlarını birbirine yaklaştırarak onları Bloch küresinin merkezine (tamamen belirsizlik noktasına) doğru sürükler. Doğa, çevreyle etkileşime giren kuantum sistemlerindeki özgün farklılıkları "zımparalayarak" birbirine benzetir.

### 4.3. Ölçümün Trajedisi (Operasyonel Sınır)
Operasyonel açıdan iz mesafesi, bir gözlemcinin veya bir kuantum sensörünün "gerçeği bilme" kapasitesidir. Helstrom Sınırı bize şunu dikte eder: 
Sistem ne kadar gürültüye maruz kalırsa, elinizdeki ölçüm aletleri ne kadar kusursuz olursa olsun, durumları doğru sınıflandırma veya ayırt etme yeteneğiniz geri döndürülemez şekilde azalır. Gürültü arttıkça, en gelişmiş kuantum sensörü bile giderek körleşir ve başarısı rastgele bir "yazı-tura" tahmini olan %50 sınırına doğru çöker.

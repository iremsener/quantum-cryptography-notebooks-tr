## 3.1.2.1 Saf durumlarda örtüşme ve fidelity

Normalize saf kuantum durumları $|\psi\rangle,|\phi\rangle\in\mathcal H$ için

$$
\langle\psi|\psi\rangle=\langle\phi|\phi\rangle=1.
$$

Bir kubit durumu

$$
|\psi\rangle=\alpha|0\rangle+\beta|1\rangle,\qquad |\alpha|^2+|\beta|^2=1
$$

biçimindedir. İki durumun **örtüşme genliği** (overlap amplitude) $\langle\psi|\phi\rangle$ karmaşık sayı olabilir. Bunun mutlak değer karesi ise olasılık olarak yorumlanabilir:

$$
\boxed{F(|\psi\rangle,|\phi\rangle)=|\langle\psi|\phi\rangle|^2.}
$$

Bu, $|\psi\rangle$ durumuna hazırlanmış sistemde $|\phi\rangle\langle\phi|$ projektörüne karşılık gelen ölçüm sonucunun olasılığıdır. $0\leq F\leq1$:

- $F=1$: Aynı fiziksel saf durum (global faz farkına izin verilir).
- $0<F<1$: Ortogonal olmayan durumlar.
- $F=0$: Ortogonal ve uygun ölçümle kusursuz ayırt edilebilir saf durumlar.

### Çözümlü örnek: $|0\rangle$ ve $|+\rangle$

$$
|0\rangle=\begin{pmatrix}1\\0\end{pmatrix},\qquad
|+\rangle=\frac{|0\rangle+|1\rangle}{\sqrt2}
=\frac1{\sqrt2}\begin{pmatrix}1\\1\end{pmatrix}.
$$

$$
\langle0|+\rangle=\frac1{\sqrt2},\qquad
\boxed{F(|0\rangle,|+\rangle)=\frac12.}
$$

Bu sonuç, $|+\rangle$ durumunun $Z$ bazında ölçülmesinde $0$ sonucunun $1/2$ olasılıkla elde edildiğini gösterir.

## 3.1.2.2 Bloch küresi geometrisi

Genel saf kubit:

$$
|\psi\rangle=\cos\frac\theta2|0\rangle+e^{i\varphi}\sin\frac\theta2|1\rangle.
$$

Saf durumların birim Bloch vektörleri $\mathbf r,\mathbf s$ arasındaki açı $\gamma$ olsun. O hâlde

$$
F=\frac{1+\mathbf r\cdot\mathbf s}{2}
=\frac{1+\cos\gamma}{2}
=\boxed{\cos^2\frac\gamma2}.
$$

Örneğin $\gamma=90^\circ$ için $F=1/2$, $\gamma=180^\circ$ için $F=0$ bulunur.

## 3.1.2.3 Karma durumlar ve Uhlmann fidelity

Karma kuantum durumları yoğunluk matrisiyle temsil edilir:

$$
\rho=\sum_i p_i|\psi_i\rangle\langle\psi_i|,\quad
p_i\ge0,\quad\sum_i p_i=1.
$$

Bu bölümde kullanılan karma durum fidelity tanımı:

$$
\boxed{F(\rho,\sigma)=\left(\operatorname{Tr}\sqrt{\sqrt\rho\,\sigma\,\sqrt\rho}\right)^2.}
$$

Durumlardan biri safsa, $\rho=|\psi\rangle\langle\psi|$ için

$$
\boxed{F(\rho,\sigma)=\langle\psi|\sigma|\psi\rangle.}
$$

Örnek olarak $\sigma=I/2$ ve $\rho=|0\rangle\langle0|$ için

$$
F(\rho,\sigma)=\frac12.
$$

**Fidelity saflık değildir.** Saflık ayrı bir niceliktir: $\operatorname{Tr}(\rho^2)$. İkisinin bazı örneklerde aynı sayıyı vermesi kavramları eşitlemez.

## 3.1.2.4 İz mesafesiyle bağlantı

Önceki bölümdeki iz mesafesi:

$$
D(\rho,\sigma)=\frac12\|\rho-\sigma\|_1,
\qquad \|A\|_1=\operatorname{Tr}\sqrt{A^\dagger A}.
$$

Fuchs–van de Graaf eşitsizlikleri (buradaki kareli fidelity konvansiyonu için):

$$
\boxed{1-\sqrt{F(\rho,\sigma)}\le D(\rho,\sigma)\le\sqrt{1-F(\rho,\sigma)}.}
$$

Örneğin $F=0.81$ ise $0.1\le D\le\sqrt{0.19}\approx0.4359$.

**İki saf durum** için üst sınır eşitliktir:

$$
\boxed{D(|\psi\rangle,|\phi\rangle)=\sqrt{1-F(|\psi\rangle,|\phi\rangle)}.}
$$

Karma durumlarda bu eşitliği genel olarak kullanmak yanlıştır.

## 3.1.2.5 Helstrom sınırı: Operasyonel ayırt edilebilirlik

Eşit önsel olasılıkla seçilmiş $\rho_0$ ve $\rho_1$ durumları için optimum tek-kopya doğru tahmin olasılığı:

$$
\boxed{P_{\mathrm{guess}}^{\mathrm{opt}}=\frac12\left(1+D(\rho_0,\rho_1)\right).}
$$

İki saf durum için:

$$
\boxed{P_{\mathrm{guess}}^{\mathrm{opt}}=\frac12\left(1+\sqrt{1-F}\right).}
$$

$|0\rangle$ ile $|+\rangle$ için:

$$
F=\frac12,\quad D=\frac1{\sqrt2},\quad
P_{\mathrm{guess}}^{\mathrm{opt}}=\frac12\left(1+\frac1{\sqrt2}\right)\approx0.8536.
$$

Bu **yalnızca belirtilen iki durumun** eşit olasılıklı ikili ayırt edilmesi içindir; BB84 protokolünün genel kırılma olasılığı değildir.

## 3.1.2.6 Qiskit deney 1 — BB84 durumları ve fidelity matrisi

BB84 durum kümesi:

$$
\mathcal S_{\mathrm{BB84}}=\{|0\rangle,|1\rangle,|+\rangle,|-\rangle\},
\qquad
|-\rangle=\frac{|0\rangle-|1\rangle}{\sqrt2}.
$$

İlk hücreyi Colab'da gerektiğinde çalıştırın:

```python
%pip -q install 'qiskit>=1.0' numpy matplotlib
```

```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, DensityMatrix, state_fidelity

circuits = {}
for name in ["0", "1", "+", "-"]:
    qc = QuantumCircuit(1)
    if name in ("1", "-"):
        qc.x(0)
    if name in ("+", "-"):
        qc.h(0)
    circuits[name] = qc

states = {name: Statevector.from_instruction(qc)
          for name, qc in circuits.items()}
labels = list(states)
F_matrix = np.array([[state_fidelity(states[a], states[b])
                      for b in labels] for a in labels])
print("Durum sırası:", labels)
print(np.round(F_matrix, 4))

fig, ax = plt.subplots(figsize=(5.5, 4.5))
im = ax.imshow(F_matrix, vmin=0, vmax=1, cmap="viridis")
ax.set_xticks(range(4), labels=[f"|{x}>" for x in labels])
ax.set_yticks(range(4), labels=[f"|{x}>" for x in labels])
for i in range(4):
    for j in range(4):
        ax.text(j, i, f"{F_matrix[i,j]:.1f}", ha="center", va="center",
                color="white" if F_matrix[i,j] < .7 else "black")
ax.set_title("BB84 durumları — fidelity matrisi")
fig.colorbar(im, ax=ax, label="Fidelity")
plt.tight_layout()
plt.show()
```

Beklenen matris:

$$
F_{ij}=\begin{pmatrix}
1&0&0.5&0.5\\
0&1&0.5&0.5\\
0.5&0.5&1&0\\
0.5&0.5&0&1
\end{pmatrix}.
$$

Aynı bazın farklı bitleri ortogonaldir; farklı BB84 bazlarının durum çiftlerinde $F=1/2$ bulunur. Bazlar **karşılıklı yansızdır** (*mutually unbiased*).

## 3.1.2.7 Qiskit deney 2 — İz mesafesini doğrulama

Hermityen $A=\rho-\sigma$ için iz normu özdeğerlerden hesaplanır:

$$
D(\rho,\sigma)=\frac12\sum_k|\lambda_k(\rho-\sigma)|.
$$

```python
def trace_distance(rho, sigma):
    delta = np.asarray(rho.data - sigma.data)
    return float(0.5 * np.abs(np.linalg.eigvalsh(delta)).sum())

for a, b in [("0", "1"), ("0", "+"), ("0", "0")]:
    rho, sigma = DensityMatrix(states[a]), DensityMatrix(states[b])
    f = state_fidelity(rho, sigma)
    d = trace_distance(rho, sigma)
    print(f"|{a}> vs |{b}>: F={f:.4f}; D={d:.4f}; sqrt(1-F)={np.sqrt(max(0,1-f)):.4f}")
```

Beklenen değerler:

| Durum çifti | Fidelity | İz mesafesi |
|---|---:|---:|
| $|0\rangle,|1\rangle$ | $0$ | $1$ |
| $|0\rangle,|+\rangle$ | $0.5$ | $\approx0.7071$ |
| $|0\rangle,|0\rangle$ | $1$ | $0$ |

## 3.1.2.8 BB84, intercept–resend ve QBER

Alice rastgele $Z$ ya da $X$ bazı seçerek kubit gönderir. Bob rastgele bazda ölçer; daha sonra kimliği doğrulanmış klasik kanalda bazlarını karşılaştırır ve eşleşmeyen turları elerler.

**Basit intercept–resend saldırısı:** Eve her kubiti yakalar, rastgele bazda ölçer, elde ettiği bite karşılık gelen durumu yeniden hazırlar ve Bob'a gönderir. İdeal, kayıpsız ve tüm sinyallerin saldırıya uğradığı modelde, elenmiş turlarda:

$$
\begin{aligned}
QBER&=P(\text{Eve yanlış baz})\,
P(\text{Bob yanlış bit}\mid\text{Eve yanlış baz})\\
&=\frac12\cdot\frac12=\boxed{\frac14=25\%}.
\end{aligned}
$$

Eve yalnızca $f$ oranındaki sinyallere saldırırsa ve başka hata kaynağı yoksa beklenen $QBER=f/4$ olur.

**Kritik ayrım:** QBER tek başına anahtar güvenliğini kanıtlamaz. Gizlilik güçlendirme, hata düzeltme, sonlu anahtar analizi ve Eve'in yan bilgisi gerekir.

## 3.1.2.9 Kanal gürültüsü, hedef fidelity ve ölçüm hatası

Depolarizasyon modelini şu parametreleştirmeyle tanımlayalım:

$$
\mathcal E_p(\rho)=(1-p)\rho+p\frac{I}{2},\qquad0\le p\le1.
$$

Hedef durum safsa $\rho=|\psi\rangle\langle\psi|$:

$$
\begin{aligned}
F\big(\rho,\mathcal E_p(\rho)\big)
&=\langle\psi|\mathcal E_p(\rho)|\psi\rangle\\
&=1-\frac p2.
\end{aligned}
$$

Bob hedefin ortonormal bazında ölçüyorsa:

$$
P_{\mathrm{error}}=1-F=\frac p2.
$$

Bunun **tek saf hedef durumu ve doğru bazdaki ideal projektif ölçüm** için geçerli olduğunu unutmayın.

## 3.1.2.10 Qiskit deney 3 — Gürültüye karşı fidelity ve iz mesafesi

```python
psi = Statevector.from_label("+")
rho = DensityMatrix(psi)
identity_half = np.eye(2, dtype=complex) / 2
ps = np.linspace(0, 1, 101)
fidelities, distances = [], []

for p in ps:
    sigma = DensityMatrix((1 - p) * rho.data + p * identity_half)
    fidelities.append(state_fidelity(rho, sigma))
    distances.append(trace_distance(rho, sigma))

plt.figure(figsize=(8, 4.5))
plt.plot(ps, fidelities, label="Fidelity")
plt.plot(ps, distances, label="Trace distance")
plt.plot(ps, 1-ps/2, "--", label="Teorik F=1-p/2")
plt.plot(ps, ps/2, ":", label="Teorik D=p/2")
plt.xlabel("Depolarizasyon parametresi p")
plt.ylabel("Değer")
plt.title("Depolarizasyon ve durum benzerliği")
plt.legend()
plt.grid(alpha=.3)
plt.tight_layout()
plt.show()
```

Bu **özel** modelde:

$$
\boxed{F=1-\frac p2,\qquad D=\frac p2,\qquad F+D=1.}
$$

Son eşitlik genel tüm kuantum durum çiftleri için geçerli değildir.

## 3.1.2.11 Qiskit deney 4 — BB84 intercept–resend simülasyonu

Aşağıdaki hücre Alice, Eve ve Bob'un durum hazırlama ve ideal projektif ölçüm adımlarını doğrudan durum vektörleri üzerinde simüle eder.

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

rng = np.random.default_rng(42)
N = 10_000
alice_bits = rng.integers(0, 2, N)
alice_bases = rng.integers(0, 2, N)
bob_bases = rng.integers(0, 2, N)
eve_bases = rng.integers(0, 2, N)

def prepare_state(bit, basis):
    """0: Z bazı; 1: X bazı."""
    qc = QuantumCircuit(1)
    if bit:
        qc.x(0)
    if basis:
        qc.h(0)
    return Statevector.from_instruction(qc)

def measure_state(state, basis):
    if basis:
        qc = QuantumCircuit(1)
        qc.h(0)
        state = state.evolve(qc)
    probabilities = state.probabilities()
    return int(rng.choice([0, 1], p=probabilities))

def run_bb84(with_eve=False):
    bob_bits = np.zeros(N, dtype=int)
    for i in range(N):
        state = prepare_state(alice_bits[i], alice_bases[i])
        if with_eve:
            eve_bit = measure_state(state, eve_bases[i])
            state = prepare_state(eve_bit, eve_bases[i])
        bob_bits[i] = measure_state(state, bob_bases[i])
    sifted = alice_bases == bob_bases
    return np.mean(alice_bits[sifted] != bob_bits[sifted]), int(sifted.sum())

clean_qber, count = run_bb84(False)
attack_qber, _ = run_bb84(True)
print(f"Elenen bit sayısı: {count}")
print(f"Saldırısız QBER: %{100*clean_qber:.2f}")
print(f"Intercept–resend QBER: %{100*attack_qber:.2f}")
```

Beklenti: İdeal saldırısız kanalda $QBER=0$; tüm sinyallere rastgele bazlı intercept–resend saldırısında büyük örneklemde $QBER\approx25\%$. Sonlu örneklemde sapmalar doğaldır.

## 3.1.2.12 CPTP kanalları ve veri işleme ilkesi

Deterministik fiziksel bir kuantum kanalı tamamen pozitif ve iz koruyan (CPTP) dönüşümdür. Kraus gösterimi:

$$
\mathcal E(\rho)=\sum_kK_k\rho K_k^\dagger,\qquad
\sum_kK_k^\dagger K_k=I.
$$

Aynı kanal her iki duruma uygulandığında:

$$
\boxed{F(\mathcal E(\rho),\mathcal E(\sigma))\ge F(\rho,\sigma),}
$$

$$
\boxed{D(\mathcal E(\rho),\mathcal E(\sigma))\le D(\rho,\sigma).}
$$

Yani aynı CPTP işleme tabi tutulan durumlar operasyonel olarak daha kolay ayırt edilemez. **Bu, kanaldaki gürültünün QKD güvenliğini otomatik olarak artırdığı anlamına gelmez:** Eve kanalın çevre sistemini kontrol ediyor veya onunla korelasyon taşıyor olabilir.

## 3.1.2.14 Özet

| Kavram | İfade / sonuç |
|---|---|
| Saf durum fidelity | $F=|\langle\psi|\phi\rangle|^2$ |
| Karma durum fidelity | $F=(\operatorname{Tr}\sqrt{\sqrt\rho\sigma\sqrt\rho})^2$ |
| İz mesafesi | $D=\frac12\|\rho-\sigma\|_1$ |
| Fuchs–van de Graaf | $1-\sqrt F\le D\le\sqrt{1-F}$ |
| Saf durumlarda | $D=\sqrt{1-F}$ |
| Helstrom (eşit önseller) | $P_{\rm guess}^{\rm opt}=(1+D)/2$ |
| Depolarizasyon (saf hedef) | $F=1-p/2$, $D=p/2$ |
| İdeal tam intercept–resend | $QBER=25\%$ |

Fidelity iki durumu ne ölçüde benzer bulduğumuzu, trace distance ne ölçüde ayırt edebildiğimizi ifade eder. QKD'de bu iki nicelik, baz uyumsuzluğu, gürültü ve saldırı modellerini açıklamaya yardım eder; ancak tek başlarına tam kriptografik güvenlik kanıtı değildir.
